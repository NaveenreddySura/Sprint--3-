# Selecting & Filtering Data 
Selecting and filtering data in Python means choosing specific rows and columns from a DataFrame based on positions, labels, or conditions. These operations are important in Data Analysis and Machine Learning because they help us extract relevant data, remove unwanted records, and prepare features for model training.

```python
%mamba install pandas
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Manoj"],
    "Age": [25, 30, 28, 24, 35, 27],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [30000, 45000, 40000, 28000, 60000, 35000]
}

df = pd.DataFrame(data)

print(df)
```

## loc

The loc property selects rows and columns from a DataFrame using their labels. It can also select records based on Boolean conditions, making it useful for label-based data selection and filtering.
### Formula 
df.loc[row_label, column_label]

- row_label specifies the row label or labels to select.
- column_label specifies the column name or names to select.
### Numerical example
If the DataFrame has index labels from 0 to 5, then df.loc[2, "Name"] returns the name associated with index label 2, which is Arun.
### When it is used
- To select specific rows by their index labels.
- To select particular columns from selected rows.
- To filter rows based on conditions.
### AI/ML example
When preparing an employee attrition dataset, you can use loc to select the Age, Salary, and Department features for analysis.

```python
# Select one row using its label
print(df.loc[2])

# Select a specific column from one row
print(df.loc[2, "Name"])

# Select multiple columns
print(df.loc[:, ["Name", "Salary"]])

# Select rows based on a condition
print(df.loc[df["Salary"] > 40000])
```

We used loc to retrieve information by index labels and column names. The expression df.loc[2, "Name"] selects the name from the row labeled 2, while df.loc[:, ["Name", "Salary"]] selects the two specified columns for all rows. In the final statement, we applied a salary condition and retrieved only employees earning more than ₹40,000. Notice that loc uses labels rather than positional row numbers.

## iloc

The iloc property selects rows and columns based on their integer positions in a DataFrame. Unlike loc, it uses zero-based positions rather than index labels or column names.
Formula / Syntax
df.iloc[row_position, column_position]

### Numerical example
For a DataFrame containing six rows, the first row has position 0, the second has position 1, and the third has position 2.
Therefore, df.iloc[2, 0] returns Arun, because the third row and first column contain that value.
### When it is used
- To select rows by their positions.
- To extract particular columns using column positions.
- To retrieve a specific section of a dataset.
### AI/ML example
When examining a transformed feature matrix, iloc can help you inspect the first 10 records or select a range of rows for debugging.

```python
# Select the third row
print(df.iloc[2])

# Select the value at the third row and first column
print(df.iloc[2, 0])

# Select the first three rows
print(df.iloc[0:3])

# Select the first two columns
print(df.iloc[:, 0:2])
```

We used iloc to select data by integer position. The expression df.iloc[2] returns the third row, while df.iloc[2, 0] returns the value in the third row and first column. The slice df.iloc[0:3] selects the first three rows because the ending position is excluded. Similarly, df.iloc[:, 0:2] selects the first two columns for every row.

## Boolean Indexing

Boolean indexing selects rows or columns by evaluating a condition that produces True or False values. Only rows whose conditions evaluate to True are retained in the filtered result.
Formula / Syntax
df[condition]

For multiple conditions:
df[(condition1) & (condition2)]

- & means AND.
- | means OR.
- ~ means NOT.
Use parentheses around individual conditions when combining them.
### Numerical example
Suppose employee salaries are:
[30000, 45000, 40000, 28000]
For the condition Salary > 35000, the Boolean result is:
[False, True, True, False]
Only the second and third records satisfy the condition.
### When it is used
- To filter records using one or more conditions.
- To select data that meets business requirements.
- To identify records that need further analysis.
### AI/ML example
In a loan approval dataset, Boolean indexing can select applicants whose income exceeds a threshold and whose credit score meets a minimum requirement.

```python
# Employees earning more than 35000
print(df[df["Salary"] > 35000])

# Employees aged above 25 AND earning above 35000
print(df[(df["Age"] > 25) & (df["Salary"] > 35000)])

# Employees belonging to IT OR HR
print(df[(df["Department"] == "IT") |
         (df["Department"] == "HR")])
```

We used Boolean indexing to filter employees based on salary, age, and department. The first condition selects employees earning more than ₹35,000. The second combines two conditions using &, so an employee must be older than 25 and earn more than ₹35,000. The third uses | to select employees who belong to either the IT or HR department. These conditions allow us to extract only the records relevant to a particular analysis.

## Filtering Rows

Filtering rows means selecting records from a DataFrame that satisfy specified conditions. It is commonly performed using comparison operators, logical operators, and Boolean masks.
### Formula 
df[df["column"] operator value]



Common comparison operators:
- > greater than
- < less than
- >= greater than or equal to
- <= less than or equal to
- == equal to
- != not equal to
### Numerical example
Suppose the ages are:
[25, 30, 28, 24, 35, 27]
Filtering with Age >= 30 selects ages 30 and 35, so two rows satisfy the condition.
### When it is used
- To remove irrelevant records from an analysis.
- To select records that satisfy business rules.
- To create subsets for reporting and modelling.
### AI/ML example
In a customer churn dataset, you can filter customers with monthly charges above a certain amount to investigate whether high charges are associated with churn.

```python
# Employees aged 28 or older
print(df[df["Age"] >= 28])

# Employees not belonging to IT
print(df[df["Department"] != "IT"])

# Employees earning between 30000 and 45000, inclusive
print(df[(df["Salary"] >= 30000) &
         (df["Salary"] <= 45000)])
```

We filtered the employee DataFrame using different comparison operators. The first statement returns employees aged 28 or older, the second excludes employees from the IT department, and the third selects employees whose salaries fall between ₹30,000 and ₹45,000, including both boundaries. Filtering rows helps us focus on the subset of records needed for a particular business question or machine learning task.

## query()

The query() method filters rows using a string expression that describes the required conditions. It provides a readable alternative to Boolean indexing, especially when filtering with multiple conditions.
### Formula 
df.query("condition")
For example:
df.query("Age > 25 and Salary > 35000")
### Numerical example
For the condition Age > 25 and Salary > 35000, an employee must satisfy both comparisons. In our dataset, Priya, Arun, and Kiran meet both conditions.
### When it is used
- To write readable filtering expressions.
- To combine multiple conditions in a compact statement.
- To make exploratory data analysis code easier to understand.
### AI/ML example
For a sales prediction model, you can use query() to extract transactions with a value above a threshold and a status of "Completed".

```python
# Employees older than 25
print(df.query("Age > 25"))

# Employees older than 25 and earning above 35000
print(df.query("Age > 25 and Salary > 35000"))

# Employees in the IT department
print(df.query('Department == "IT"'))
```

We used query() to filter employees by writing conditions as strings. The first statement selects employees older than 25, and the second combines age and salary requirements using the word and. The final statement selects employees whose department is IT. This method can make complex filters easier to read, although column names containing spaces or special characters may require special handling.

## isin()

The isin() method checks whether each value in a Series belongs to a specified collection of values. It returns a Boolean result that can be used to filter rows matching any value in that collection.
### Formula 
df["column"].isin([value1, value2, value3])
### Numerical example
Suppose the departments are:
["IT", "HR", "IT", "Finance"]
Checking membership in ["IT", "HR"] produces:
[True, True, True, False]
The Finance record does not match either requested department.
### When it is used
- To filter rows matching multiple possible values.
- To check whether values belong to an approved list.
- To avoid writing several OR conditions manually.
### AI/ML example
In a customer segmentation model, isin() can select customers from a specified list of regions or membership categories.

```python
# Select employees from IT and HR
print(df[df["Department"].isin(["IT", "HR"])])

# Select employees whose names are Ravi, Sneha, or Manoj
print(df[df["Name"].isin(["Ravi", "Sneha", "Manoj"])])
```

We used isin() to check whether each employee's department appears in the list ["IT", "HR"]. The resulting Boolean mask selects all matching rows and excludes Sneha, who belongs to Finance. We then applied the same technique to select employees whose names belong to a specified list. This is particularly useful when the filter contains many acceptable values.

## between()

The between() method checks whether values in a Series fall within a specified numerical or comparable range. By default, it includes both the lower and upper boundaries.
### Formula 
df["column"].between(lower, upper)
By default, inclusive="both" means that values equal to either boundary are included.
### Numerical example
Consider these salaries:
[28000, 30000, 35000, 40000, 60000]
For the range 30,000 to 40,000, inclusive, the matching values are:
[30000, 35000, 40000]
Three values fall within the specified range.
### When it is used
- To select values within a specified range.
- To filter age, salary, price, marks, or other numerical features.
- To make range-based conditions shorter and clearer.
### AI/ML example
In a house-price dataset, between() can select houses with areas between 1,000 and 2,000 square feet for a focused analysis.

```python
# Employees with salaries between 30000 and 45000
print(df[df["Salary"].between(30000, 45000)])

# Employees aged between 25 and 30
print(df[df["Age"].between(25, 30)])
```

We used between() to select employees whose salaries fall between ₹30,000 and ₹45,000, including both endpoints. The expression returns a Boolean mask, which we pass to the DataFrame to retrieve only the matching rows. The second statement applies the same approach to ages between 25 and 30. This method simplifies range-based filtering compared with writing two separate comparison conditions.

## Sorting Data

Sorting arranges DataFrame rows according to the values in one or more columns. Pandas provides sort_values() for sorting by column values and sort_index() for sorting by index labels.
### Formula 
df.sort_values(by="column", ascending=True)
- by specifies the column to sort.
- ascending=True sorts from smallest to largest or alphabetically.
- ascending=False sorts from largest to smallest or in reverse alphabetical order.
### Numerical example
Suppose salaries are:
[30000, 45000, 40000, 28000]
Ascending order:
[28000, 30000, 40000, 45000]
Descending order:
[45000, 40000, 30000, 28000]
### When it is used
- To rank records by salary, price, marks, or sales.
- To identify the highest or lowest values.
- To organize data before reporting or analysis.
### AI/ML example
In a recommendation system, you can sort products by predicted rating to inspect the highest-rated recommendations first. Sorting is for ordering the data; it does not itself train or improve the model.

```python
# Sort by salary in ascending order
print(df.sort_values(by="Salary"))

# Sort by salary in descending order
print(df.sort_values(by="Salary", ascending=False))

# Sort by department, then salary
print(df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
))
```

We used sort_values() to arrange employee records according to salary. The first statement sorts salaries from lowest to highest, while the second reverses the order to show the highest salary first. The third statement sorts by department alphabetically and then sorts salaries in descending order within each department. By default, sorting returns a new DataFrame and does not modify the original DataFrame unless you explicitly use an in-place operation or assign the result back.

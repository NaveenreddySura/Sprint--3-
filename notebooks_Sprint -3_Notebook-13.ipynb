# Notebook 13 – Missing Values 
Missing values are data points that are absent or unknown in a dataset. In Python, pandas provides methods to identify, remove, and fill missing values so that the data is suitable for analysis and Machine Learning.

```python
%mamba install pandas
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran"],
    "Age": [25, np.nan, 28, 24, np.nan],
    "Salary": [30000, 45000, np.nan, 28000, 60000],
    "Department": ["IT", "HR", None, "Finance", "HR"]
})

print(df)
```

## isnull()

The isnull() method identifies missing values in a pandas Series or DataFrame. It returns True for missing values such as NaN and None, and False for values that are present.
### Formula 
df.isnull()
To count missing values in each column:
df.isnull().sum()
### When it is used
- To identify missing data in a dataset.
- To count missing values in individual columns.
- To decide which columns require data cleaning.
### AI/ML example
In a house-price prediction dataset, isnull() helps identify missing house sizes or prices before training a regression model.

```python
print(df.isnull())
print("\nMissing values per column:")
print(df.isnull().sum())
```

We used isnull() to check every cell in the DataFrame and identify missing values. The method returns a Boolean DataFrame in which missing entries are marked True. We then used sum() to count the missing values in each column because pandas treats True as 1 and False as 0 during summation. The output shows that Age has two missing values, Salary has one, and Department has one, while Name has no missing values.

## notnull()

The notnull() method identifies values that are present in a pandas Series or DataFrame. It returns True for non-missing values and False for missing values.
### Formula 
df.notnull()
To count non-missing values in each column:
df.notnull().sum()
### When it is used
- To identify records containing available data.
- To filter out rows with missing values in a specific column.
- To count valid observations before analysis.
### AI/ML example
In a customer churn dataset, you can use notnull() to select customers whose monthly charges are available before analyzing spending patterns.

```python
print(df["Age"].notnull())

print("\nEmployees with known ages:")
print(df[df["Age"].notnull()])
```

We used notnull() on the Age column to determine which employees have a recorded age. The method returns True for Ravi, Arun, and Sneha because their ages are available, and False for Priya and Kiran because their ages are missing. We then used this Boolean result to filter the DataFrame and retain only employees whose ages are known. This is useful when a particular analysis requires a specific field to be present.

## dropna()

The dropna() method removes rows or columns containing missing values from a DataFrame. By default, it removes any row containing at least one missing value, but its behavior can be customized.
### Formula 
df.dropna()
### Useful options:
- axis=0: Remove rows.
- axis=1: Remove columns.
- how="all": Remove rows or columns only when all their values are missing.
- subset=["Age"]: Check missing values only in specified columns.
### When it is used
- To remove incomplete records when appropriate.
- To exclude rows missing essential information.
- To remove columns that contain only missing values.
### AI/ML example
In a model predicting house prices, you may remove rows where the target price is missing because those records cannot be used as labeled examples for supervised training.

```python
df_dropped = df.dropna()

print(df_dropped)
```

We used dropna() to remove every row containing at least one missing value. Only Ravi and Sneha remain because all their fields are present. In the second example, subset=["Age"] restricts the check to the Age column, so employees with missing salaries or departments can still be retained as long as their ages are available. Removing rows is simple, but it can discard useful information and should be based on the importance and amount of missing data.

## fillna()

The fillna() method replaces missing values with a specified value or values calculated from the available data. It is one of the most common techniques for handling incomplete datasets without deleting entire records.
### Formula
For filling missing numerical values using the mean:
\[
\text{Mean}=\frac{\sum x_i}{n}
\]

Here, \(x_i\) represents each available value and \(n\) is the number of non-missing values.
### Numerical example
Consider the available ages:
25, 28, 24
The mean age is:
\[
\frac{25+28+24}{3}=\frac{77}{3}\approx25.67
\]

A missing age can be filled with approximately 25.67 using mean imputation.
### When it is used
- To fill missing numerical values with mean or median.
- To replace missing categorical values with a common category.
- To preserve rows that would otherwise be removed.
### AI/ML example
In a customer churn dataset, missing numerical features such as monthly charges may be filled using the training-set median, while missing categorical features may be filled using the most frequent category.

```python
df_filled = df.copy()

# Fill missing ages with the mean age
df_filled["Age"] = df_filled["Age"].fillna(
    df_filled["Age"].mean()
)

# Fill missing salary with the median salary
df_filled["Salary"] = df_filled["Salary"].fillna(
    df_filled["Salary"].median()
)

# Fill missing department with a label
df_filled["Department"] = df_filled["Department"].fillna(
    "Unknown"
)

print(df_filled)
```

We first copied the original DataFrame so that the source data remained unchanged. We then filled missing ages using the mean of the available ages, missing salaries using the median salary, and the missing department using the label "Unknown". The mean is useful for numerical data that is reasonably balanced, while the median is more resistant to extreme values. Using "Unknown" preserves the fact that the department was not recorded rather than assuming a particular department.

## interpolate()

The interpolate() method estimates missing values using values surrounding the gaps or another supported interpolation method. It is especially useful for ordered numerical data, such as time-series measurements, where neighboring observations provide meaningful information.
### Formula
For linear interpolation between two known values:
\[
y=y_1+\frac{x-x_1}{x_2-x_1}(y_2-y_1)
\]

### Numerical example
Suppose a sensor records:
10, NaN, 20
If the missing value is halfway between the two observations, linear interpolation estimates it as:
\[
10+\frac{1}{2}(20-10)=15
\]

### When it is used
- To fill gaps in time-series data.
- To estimate missing measurements between known observations.
- To preserve trends in ordered numerical data.
### AI/ML example
In a weather forecasting project, interpolation may estimate a missing temperature reading between two known measurements, provided that the gap is small and the assumption of a gradual change is reasonable.

```python
df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Temperature": [20.0, 22.0, None, 26.0, 28.0]
})

df["Temperature_Filled"] = df["Temperature"].interpolate(
    method="linear"
)

print(df)
```

We created a temperature dataset with one missing reading on Day 3. Using linear interpolation, pandas estimates the missing temperature from the neighboring values of 22°C and 26°C, resulting in 24°C. We stored the estimated values in a separate column so that the original measurements remained available. This technique works well when observations are ordered and gradual changes are plausible, but it may be unsuitable for categorical data, abrupt changes, or large gaps.

## Missing Value Handling Strategies

Missing value handling strategies are methods used to decide how incomplete records should be treated before data analysis or model training. The appropriate strategy depends on why values are missing, how much data is affected, and whether missingness itself contains useful information.
### Formula
No single formula applies to all strategies. For example, the missing-value percentage can be calculated as:
\[
\text{Missing Percentage}=
\frac{\text{Missing Values}}{\text{Total Values}}\times100
\]

### Numerical example
Suppose a dataset contains 200 records, and the Age column has 30 missing values.
\[
\frac{30}{200}\times100=15\%
\]

Therefore, 15% of the Age values are missing.
### When it is used
- To choose a suitable cleaning strategy before analysis.
- To preserve useful data while minimizing bias.
- To prepare reliable training and evaluation datasets.
### AI/ML example
For a house-price prediction model, you might use median imputation for missing numerical features, a separate "Unknown" category for missing categorical values, and remove rows where the target price is unavailable for supervised training.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, np.nan, 30, 35, np.nan],
    "Salary": [30000, 40000, np.nan, 60000, 50000],
    "Department": ["IT", "HR", None, "IT", "Finance"]
})

# Calculate missing percentages
missing_percentage = df.isnull().mean() * 100

print("Missing percentage:")
print(missing_percentage)

# Create indicators before filling
df["Age_Was_Missing"] = df["Age"].isnull().astype(int)

# Median imputation for Age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Mean imputation for Salary
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Constant imputation for Department
df["Department"] = df["Department"].fillna("Unknown")

print("\nCleaned dataset:")
print(df)
```

We first calculated the percentage of missing values in each column using isnull().mean() * 100. We then created an indicator column to record which ages were originally missing, filled missing ages using the median, filled missing salaries using the mean, and replaced the missing department with "Unknown". The resulting dataset has no missing values in these three original columns, while the indicator retains information about missing ages. For Machine Learning, imputation values should generally be learned from the training data only and then applied to validation and test data to avoid data leakage.

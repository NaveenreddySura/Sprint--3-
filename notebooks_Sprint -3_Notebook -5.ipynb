# Reading & Writing Data

## CSV Files

CSV (Comma-Separated Values) is a plain-text file format that stores tabular data in rows and columns, usually separated by commas. It is widely supported by spreadsheets, databases, and data-processing applications.
### Where It Is Used
- Sharing tabular data between applications.
- Exporting database query results.
- Storing datasets for analysis.
- Exchanging data between Python and spreadsheet tools.
### AI/ML Engineer Example
An AI/ML engineer reads a customer dataset from a CSV file, cleans missing values, and saves the processed dataset for model training.

```python
import pandas as pd
import sqlite3
from pathlib import Path

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Kiran"],
    "Age": [25, 28, 24],
    "Salary": [30000, 45000, 35000]
})

output_dir = Path("data_files")
output_dir.mkdir(parents=True, exist_ok=True)

print(df)
```

## JSON Files

JSON (JavaScript Object Notation) is a text-based format that stores structured data using objects, arrays, and key-value pairs. It is commonly used for APIs, web applications, and exchanging data between different systems.
### Where It Is Used
- Reading responses from web APIs.
- Storing structured application data.
- Exchanging data between Python and web applications.
- Saving configuration and nested data.
### AI/ML Engineer Example
An AI/ML engineer receives predictions from a REST API in JSON format, converts them into a DataFrame for analysis, and exports processed results for another application.

```python
import pandas as pd
from pathlib import Path

file_path = Path("data_files/employees.json")

# Write DataFrame to JSON
df.to_json(file_path, orient="records", indent=4)

# Read JSON into a DataFrame
json_df = pd.read_json(file_path)

print("JSON Data:")
print(json_df)
```

We used df.to_json() to convert the DataFrame into JSON and save it to a file. The orient="records" argument stores each row as a separate JSON object, while indent=4 makes the file easier to read. We then used pd.read_json() to load the records back into a DataFrame. This workflow is useful when exchanging structured data with APIs and other applications.
### Useful functions:
- pd.read_json() — reads JSON data.
- df.to_json() — writes a DataFrame to JSON.
- df.to_json(orient="records", indent=4) — writes formatted JSON records.

## Parquet Files

Parquet is a column-oriented binary file format designed for efficient storage and analytical processing of large datasets. It typically provides compression and efficient column-based reading while preserving data types more effectively than plain CSV.
### Where It Is Used
- Big data analytics and data engineering.
- Data lakes and cloud storage.
- Apache Spark processing.
- Efficient storage of machine learning datasets.
### AI/ML Engineer Example
An AI/ML engineer loads only the features required for training from a large Parquet dataset, reducing unnecessary data loading and improving preprocessing efficiency.

```python
%mamba install pyarrow
```

```python
from pathlib import Path

file_path = Path("data_files/employees.parquet")

# Write DataFrame to Parquet
df.to_parquet(file_path, index=False)

# Read the Parquet file
parquet_df = pd.read_parquet(file_path)

print("Parquet Data:")
print(parquet_df)

# Read selected columns only
selected_df = pd.read_parquet(
    file_path,
    columns=["Name", "Salary"]
)

print("\nSelected Columns:")
print(selected_df)
```

We used df.to_parquet() to save the employee DataFrame in Parquet format, with index=False preventing the row index from being stored as a separate column. We then used pd.read_parquet() to load the entire file into a new DataFrame. Finally, we used the columns argument to read only the Name and Salary columns. This ability to read selected columns is particularly useful when working with large analytical datasets.
### Useful functions:
- pd.read_parquet() — reads Parquet data.
- df.to_parquet() — writes Parquet data.
- pd.read_parquet(file_path, columns=["Name", "Salary"]) — reads selected columns.

## Pickle Files

Pickle is a Python serialization format that saves Python objects in a binary file so they can be restored later. Pandas supports Pickle for storing and retrieving DataFrames while preserving much of their original structure and data types.
### Where It Is Used
- Saving intermediate Python objects.
- Reusing processed DataFrames.
- Caching data during development.
- Storing objects in Python-only workflows.
### AI/ML Engineer Example
An AI/ML engineer saves a processed DataFrame as a Pickle file to avoid repeating expensive preprocessing steps during experimentation.

```python
import pandas as pd
from pathlib import Path

file_path = Path("data_files/employees.pkl")

# Write DataFrame to Pickle
df.to_pickle(file_path)

# Read the Pickle file
pickle_df = pd.read_pickle(file_path)

print("Pickle Data:")
print(pickle_df)
```

We used df.to_pickle() to serialize the DataFrame into a binary file. We then used pd.read_pickle() to restore the saved object into pickle_df and printed it. Pickle is convenient when saving intermediate results in Python, but it is not a general-purpose data interchange format. Never load Pickle files from untrusted sources, because doing so can execute malicious code.
### Useful functions:
- df.to_pickle() — saves a Python object.
- pd.read_pickle() — restores a saved object.

## SQL Tables

SQL tables store structured information in relational databases, with data organized into rows and columns. Pandas can execute SQL queries and load their results into a DataFrame, or write DataFrame contents into database tables.
### Where It Is Used
- Reading records from relational databases.
- Filtering data with SQL queries.
- Saving processed data to database tables.
- Building data pipelines between Python and business systems.
### AI/ML Engineer Example
An AI/ML engineer retrieves customer features from a database, processes them with Pandas, and saves the prepared data to a table for downstream model training.
Python Implementation — Reading and Writing
We will use SQLite, which is included with Python and does not require a separate database server.

```python
import pandas as pd
import sqlite3
from pathlib import Path

db_path = Path("data_files/employees.db")

with sqlite3.connect(db_path) as conn:

    # Write DataFrame to a SQL table
    df.to_sql(
        "employees",
        conn,
        if_exists="replace",
        index=False
    )

    # Read data using a SQL query
    sql_df = pd.read_sql_query(
        "SELECT * FROM employees WHERE Salary >= 35000",
        conn
    )

    print("SQL Query Results:")
    print(sql_df)

    # Read the entire table
    all_employees = pd.read_sql_query(
        "SELECT * FROM employees",
        conn
    )

    print("\nAll Employees:")
    print(all_employees)
```

We connected to an SQLite database file using sqlite3.connect(). The df.to_sql() method created or replaced the employees table and inserted the DataFrame records into it. We then used pd.read_sql_query() to execute a query that selected employees whose salaries were at least 35000, followed by another query that retrieved the entire table. The results of both queries were returned as DataFrames. The with statement manages the connection and closes it when the block ends.
Important: if_exists="replace" replaces an existing table and its contents. Use if_exists="append" to add records to an existing table, or if_exists="fail" to raise an error if the table already exists.
### Useful functions:
- pd.read_sql_query() — reads the result of a SQL query.
- pd.read_sql_table() — reads a database table when supported by the connection and SQLAlchemy setup.
- df.to_sql() — writes a DataFrame to a SQL table.

## Reading Multiple Files

Reading multiple files means loading several files that share a format or structure and combining their contents into one dataset. Python's pathlib and glob tools help locate files, while Pandas reads and combines their data.
### Where It Is Used
- Combining monthly sales reports.
- Processing datasets split across multiple files.
- Merging records collected from different sources.
- Preparing large datasets for machine learning.
### AI/ML Engineer Example
An AI/ML engineer receives transaction data split across multiple CSV files and combines them into one DataFrame before cleaning the data and creating model features.

```python
import pandas as pd
from pathlib import Path

folder = Path("data_files/monthly")
folder.mkdir(parents=True, exist_ok=True)

# Create two sample datasets
part1 = pd.DataFrame({
    "Employee_ID": [101, 102],
    "Name": ["Ravi", "Priya"],
    "Salary": [30000, 45000]
})

part2 = pd.DataFrame({
    "Employee_ID": [103, 104],
    "Name": ["Kiran", "Anu"],
    "Salary": [35000, 40000]
})

# Write separate CSV files
part1.to_csv(folder / "employees_part1.csv", index=False)
part2.to_csv(folder / "employees_part2.csv", index=False)

# Find all matching CSV files
file_paths = sorted(folder.glob("employees_*.csv"))

# Read every file
dataframes = [
    pd.read_csv(file_path)
    for file_path in file_paths
]

# Combine all DataFrames
combined_df = pd.concat(dataframes, ignore_index=True)

print("Combined Data:")
print(combined_df)
```

We created two DataFrames and saved them as separate CSV files in the same folder. The folder.glob("employees_*.csv") expression located files whose names matched the specified pattern, and sorted() kept the file processing order consistent. We loaded each file with pd.read_csv() and combined the resulting DataFrames using pd.concat(). The ignore_index=True argument created a fresh sequential index for the combined dataset. This approach is useful for automating data pipelines that receive multiple files.
Important: Check that the files have compatible column names and data types before combining them. If columns differ, Pandas aligns them by name and fills missing values with NaN.

##File Path Handling

File path handling means specifying and managing the locations of files and folders that a Python program reads or writes. Python's pathlib library provides a cross-platform way to build paths, check file existence, create directories, and organize file operations.
### Where It Is Used
- Loading files from project directories.
- Saving processed datasets to specific locations.
- Checking whether a file exists before reading it.
- Organizing input and output folders in data pipelines.
### AI/ML Engineer Example
An AI/ML engineer organizes a project into raw, processed, and models folders so that preprocessing scripts can locate source data and save cleaned datasets in the correct locations.

```python
import pandas as pd
from pathlib import Path

# Create project directories
data_dir = Path("data_files")
raw_dir = data_dir / "raw"
processed_dir = data_dir / "processed"

raw_dir.mkdir(parents=True, exist_ok=True)
processed_dir.mkdir(parents=True, exist_ok=True)

# Define input and output paths
input_path = raw_dir / "employees.csv"
output_path = processed_dir / "employees_cleaned.csv"

# Create a sample input file if it does not exist
if not input_path.exists():
    df.to_csv(input_path, index=False)

# Read and save the data
if input_path.exists():
    employees = pd.read_csv(input_path)

    employees.to_csv(output_path, index=False)

    print("Input Exists:", input_path.exists())
    print("Output Exists:", output_path.exists())
    print("Input Path:", input_path.resolve())
    print("Output Path:", output_path.resolve())
else:
    print("Input file not found.")
```

We used Path() to define the project directories and the / operator to combine path components in a platform-independent way. The mkdir() method created the input and output folders if they did not already exist. We checked whether the input file existed using exists() and created a sample CSV file only when necessary. After reading the input with pd.read_csv(), we saved the resulting DataFrame to the processed folder and printed the resolved file paths. This makes data-processing scripts more organized and easier to maintain across Windows, Linux, and macOS.
### Useful path operations:
Operation	Purpose

path.exists()	Check whether a file or directory exists
path.resolve()	Get the absolute path
path.parent	Get the parent directory

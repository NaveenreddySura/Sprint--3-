## Create a NumPy array

A NumPy array is a data structure used to store and process numerical data efficiently. Unlike a regular Python list, a NumPy array usually stores elements of the same data type and supports fast mathematical operations.
### When it is used
- Performing mathematical calculations.
- Storing numerical datasets.
- Preparing data for machine learning.
- Working with matrices and multidimensional data.

```python
import numpy as np

# Create a one-dimensional array
arr = np.array([10, 20, 30, 40, 50])

# Create a two-dimensional array
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("1D Array:")
print(arr)

print("\n2D Array:")
print(matrix)

print("\nShape:", matrix.shape)
print("Data type:", arr.dtype)
```

### Explanation
- np.array() converts the supplied list into a NumPy array.
- arr is a one-dimensional array containing five elements.
- matrix is a two-dimensional array with two rows and three columns.
- .shape returns the dimensions of the array.
- .dtype shows the data type of its elements. The exact integer type may vary by platform.

## Difference between a NumPy Array and a Python List
A Python list is a general-purpose container that can hold different data types and supports flexible operations. A NumPy array is designed primarily for efficient numerical and multidimensional computation.
### When it is used
- Python list: General-purpose collections, mixed data types, and flexible insertion or removal.
- NumPy array: Numerical calculations, scientific computing, and large numerical datasets.

```python
import numpy as np

# Python list
numbers_list = [10, 20, 30]

# NumPy array
numbers_array = np.array([10, 20, 30])

print("List multiplied by 2:")
print(numbers_list * 2)

print("\nArray multiplied by 2:")
print(numbers_array * 2)
```

### Explanation
- Multiplying a Python list by 2 repeats the list twice.
- Multiplying a NumPy array by 2 multiplies each element by two.
- NumPy arrays generally offer faster numerical operations and more compact storage for homogeneous numerical data.
- Lists are more flexible when elements have different types or when you need general-purpose collection behavior.

## Explain Broadcasting with an example

Broadcasting is a NumPy feature that allows arrays with compatible shapes to participate in mathematical operations without manually repeating or reshaping every value.
For example, NumPy can add a single number to every element in an array.
### When it is used
- Adding or multiplying a constant across an array.
- Normalizing numerical data.
- Applying different values to rows or columns of a matrix.
- Performing machine learning calculations efficiently.

```python
import numpy as np

arr = np.array([10, 20, 30])

result = arr + 5

print("Original array:", arr)
print("After broadcasting:", result)

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

addition = matrix + np.array([10, 20, 30])

print("\nMatrix broadcasting:")
print(addition)
```

### Explanation
- In arr + 5, NumPy adds 5 to each element without needing a loop.
- In the matrix example, the one-dimensional array [10, 20, 30] is applied to each row.
- NumPy supports this because the array of three values is compatible with the matrix's three columns.
- Broadcasting avoids manually copying values and keeps the code concise.

## Difference between Series and DataFrame

A Pandas Series is a one-dimensional labeled data structure, similar to a single column. A DataFrame is a two-dimensional labeled structure containing rows and columns.
### When it is used
- Series: Working with one column, such as salary or age.
- DataFrame: Working with a complete dataset containing multiple columns.

```python
import pandas as pd

# Create a Series
ages = pd.Series([22, 25, 30], name="Age")

# Create a DataFrame
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [22, 25, 30],
    "Salary": [30000, 40000, 50000]
})

print("Series:")
print(ages)

print("\nDataFrame:")
print(df)

print("\nAge column from DataFrame:")
print(df["Age"])
```

### Explanation
- pd.Series() creates a single labeled column of values.
- pd.DataFrame() creates a table containing multiple columns.
- df["Age"] selects the Age column and returns a Series.
- A DataFrame can contain columns with different data types, such as strings and integers.

## Read a CSV file using Pandas

A CSV (Comma-Separated Values) file stores tabular data as text, with values separated by delimiters, usually commas. Pandas provides read_csv() to load CSV data into a DataFrame.
### When it is used
- Loading datasets for analysis.
- Reading exported reports.
- Preparing data for machine learning.
- Importing employee, sales, customer, or transaction records.

```python
import pandas as pd

# Create sample data
data = {
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [22, 25, 30],
    "Salary": [30000, 40000, 50000]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("employees.csv", index=False)

# Read the CSV file
loaded_df = pd.read_csv("employees.csv")

print(loaded_df)
```

### Explanation
- pd.DataFrame() creates a table from a Python dictionary.
- to_csv() saves the table to a CSV file.
- index=False prevents Pandas from saving its row index as an extra column.
- pd.read_csv() reads the saved file back into a DataFrame.
- In real projects, you can directly read an existing file using pd.read_csv("employees.csv").

## Merge two DataFrames

Merging combines two DataFrames using one or more matching columns, similar to joining tables in SQL.
Pandas provides merge() and supports join types such as inner, left, right, and outer.
### When it is used
- Combining employee details with salary records.
- Combining customer information with orders.
- Joining datasets from different sources.
- Preparing a single dataset for analysis or machine learning.

```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Arun"]
})

salaries = pd.DataFrame({
    "Employee_ID": [101, 102, 104],
    "Salary": [30000, 40000, 50000]
})

# Inner merge
inner_result = pd.merge(
    employees,
    salaries,
    on="Employee_ID",
    how="inner"
)

print("Inner Merge:")
print(inner_result)

# Left merge
left_result = pd.merge(
    employees,
    salaries,
    on="Employee_ID",
    how="left"
)

print("\nLeft Merge:")
print(left_result)
```

### Explanation
- The first DataFrame contains employee IDs and names.
- The second DataFrame contains employee IDs and salaries.
- on="Employee_ID" specifies the common column used to match records.
- how="inner" returns only IDs present in both DataFrames.
- how="left" preserves every employee from the first DataFrame and fills unmatched salary values with NaN.

### Types of merge
Merge type	Result
inner	Only matching records
left	All left records and matching right records
right	All right records and matching left records
outer	All records from both sides

## Perform GroupBy and Aggregation

groupby() divides a DataFrame into groups based on one or more columns. Aggregation calculates summary values for each group, such as the sum, average, minimum, maximum, or count.
### When it is used
- Calculating average salary by department.
- Summarizing sales by product or region.
- Counting customers by category.
- Analyzing groups in AI/ML datasets.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "Sales"],
    "Employee": ["Ravi", "Priya", "Arun", "Meena", "Kiran"],
    "Salary": [50000, 40000, 60000, 45000, 35000]
})

# Calculate average salary
average_salary = df.groupby("Department")["Salary"].mean()

print("Average Salary:")
print(average_salary)

# Multiple aggregations
summary = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Maximum_Salary=("Salary", "max"),
    Employee_Count=("Employee", "count")
)

print("\nDepartment Summary:")
print(summary)
```

### Explanation
- groupby("Department") creates a group for each department.
- ["Salary"].mean() calculates the average salary for every group.
- .agg() calculates multiple statistics in a single operation.
- Average_Salary, Maximum_Salary, and Employee_Count are the names of the output columns.
- This approach summarizes a large dataset without manually creating separate subsets for every department.

## Filter rows based on multiple conditions

Filtering selects rows that satisfy specified conditions. In Pandas, Boolean indexing allows you to combine conditions using & (AND), | (OR), and ~ (NOT).
### When it is used
- Selecting employees above a salary threshold.
- Finding customers from a particular city and age range.
- Identifying transactions that meet multiple business rules.
- Selecting relevant records before machine learning.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena", "Kiran"],
    "Age": [22, 28, 35, 30, 24],
    "Salary": [30000, 50000, 60000, 45000, 55000],
    "Department": ["IT", "HR", "IT", "Finance", "IT"]
})

# Condition 1: Age greater than 25
# Condition 2: Salary greater than 40000
# Condition 3: Department is IT

filtered_df = df[
    (df["Age"] > 25) &
    (df["Salary"] > 40000) &
    (df["Department"] == "IT")
]

print(filtered_df)
```

### Explanation
- df["Age"] > 25 selects rows where the age is greater than 25.
- df["Salary"] > 40000 selects rows where the salary exceeds 40,000.
- df["Department"] == "IT" selects IT employees.
- The & operator requires all three conditions to be true.
- Parentheses around each condition are important because of Python operator precedence.

## Handle missing values

Missing values occur when a dataset has unavailable or unrecorded information. Pandas commonly represents missing numeric or general data using NaN or None, depending on the column's data type.
Pandas provides methods such as isnull(), dropna(), and fillna() to identify, remove, or replace missing values.
### When it is used
- Cleaning incomplete datasets.
- Preparing data before training a machine learning model.
- Handling missing customer or employee details.
- Preventing unexpected results in calculations.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Meena"],
    "Age": [22, None, 30, 28],
    "Salary": [30000, 40000, None, 50000]
})

print("Original DataFrame:")
print(df)

print("\nMissing values per column:")
print(df.isnull().sum())

# Fill missing Age with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with the mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\nAfter handling missing values:")
print(df)
```

### Explanation
- isnull().sum() counts missing values in each column.
- The median of the available ages is 28, so Priya's missing age is replaced with 28.
- The mean of the available salaries is 40,000, so Arun's missing salary is replaced with 40,000.
- fillna() returns the values with missing entries replaced.
- The appropriate strategy depends on the meaning of the data; a missing salary, for example, should not automatically be assumed to equal the average.

## Calculate statistical measures using NumPy

NumPy provides functions to calculate statistical measures that describe the distribution of numerical data. These include mean, median, standard deviation, variance, minimum, maximum, and percentiles.
### When it is used
- Understanding numerical datasets.
- Comparing student marks or employee salaries.
- Detecting unusual values.
- Performing exploratory data analysis before machine learning.
Python example

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", round(np.std(marks), 2))
print("Variance:", np.var(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("25th Percentile:", np.percentile(marks, 25))
```

### Explanation
- np.mean() calculates the average of all marks.
- np.median() returns the middle value after sorting the data.
- np.std() measures how spread out the values are around the mean.
- np.var() calculates the variance, which is the squared standard deviation.
- np.min() and np.max() return the smallest and largest values.
- np.percentile() calculates a percentile, which helps describe where values fall within a dataset.

## Optimize a DataFrame for memory usage

DataFrame memory optimization reduces the RAM required to store data by choosing appropriate data types and avoiding unnecessary columns. It is especially useful when datasets contain many rows or repeated text values.
### When it is used
- Processing large CSV files.
- Working with limited RAM.
- Building data pipelines for machine learning.
- Improving the scalability of Pandas workflows.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 30, 28, 35],
    "Salary": [30000, 45000, 60000, 50000, 55000],
    "Department": ["IT", "HR", "IT", "Finance", "HR"]
})

# Check memory before optimization
before = df.memory_usage(deep=True).sum()

print("Memory before:", before, "bytes")

# Convert to suitable data types
df["Age"] = df["Age"].astype("int8")
df["Salary"] = df["Salary"].astype("int32")
df["Department"] = df["Department"].astype("category")

# Check memory after optimization
after = df.memory_usage(deep=True).sum()

print("Memory after:", after, "bytes")
print("\nOptimized data types:")
print(df.dtypes)
```

### Explanation
- memory_usage(deep=True).sum() estimates the memory used by the DataFrame, including the contents of object columns.
- int8 uses less storage than int64 and can safely store the example ages.
- int32 uses less storage than int64 and is sufficient for the example salaries.
- category can save memory when a text column contains many repeated values.
- Comparing memory before and after shows whether the changes were beneficial. On a small DataFrame, the improvement may be minor; larger datasets can show more substantial savings.

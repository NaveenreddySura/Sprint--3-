# Notebook 12 – String Operations 
String operations in Python are used to clean, modify, search, and extract information from text data. In pandas, string methods are accessed through the .str accessor, which allows us to process text across an entire column efficiently.

```python
%mamba install pandas
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": [
        "  RAVI ",
        "Priya",
        " ARUN ",
        "Sneha",
        "KIRAN"
    ],
    "Email": [
        "RAVI@GMAIL.COM",
        "priya@yahoo.com",
        "arun@gmail.com",
        "sneha@outlook.com",
        "kiran@gmail.com"
    ],
    "Department": [
        "IT",
        "Human Resources",
        "IT",
        "Finance",
        "Human Resources"
    ]
})

print(df)
```

## lower()

The lower() method converts all uppercase letters in a string into lowercase letters. In pandas, .str.lower() applies this conversion to every string value in a Series.
### Formula 
df["column"].str.lower()

No mathematical formula is required.
### When it is used
- To standardize text with inconsistent capitalization.
- To make text comparisons case-insensitive.
- To clean text before searching or analyzing data.
### AI/ML example
In a spam email classification model, converting all email text to lowercase helps standardize words such as "OFFER", "Offer", and "offer".

```python
df["Name_Lower"] = df["Name"].str.lower()

print(df[["Name", "Name_Lower"]])
```

We used .str.lower() to convert all names into lowercase and stored the results in a new column called Name_Lower. For example, "RAVI" becomes "ravi" and "KIRAN" becomes "kiran". This standardization helps avoid treating differently capitalized versions of the same text as different values during filtering, grouping, or Machine Learning preprocessing.

## upper()

The upper() method converts all lowercase letters in a string into uppercase letters. In pandas, .str.upper() applies this transformation to every string value in a column.
### Formula 
df["column"].str.upper()
No mathematical formula is required.
### When it is used
- To standardize text into uppercase.
- To format codes and identifiers consistently.
- To make text easier to compare when capitalization is inconsistent.
### AI/ML example
In a customer database, country codes such as "in", "In", and "IN" can be standardized to "IN" before grouping customers by country.

```python
df["Department_Upper"] = df["Department"].str.upper()

print(df[["Department", "Department_Upper"]])
```

We used .str.upper() to convert each department value into uppercase and stored the result in Department_Upper. Values such as "Finance" become "FINANCE", while "IT" remains unchanged because it is already uppercase. This is useful for standardizing category names and preparing text for consistent comparisons across different data sources.

## strip()

The strip() method removes unwanted whitespace or specified characters from the beginning and end of a string. In pandas, .str.strip() applies this operation to every string value in a column without removing spaces inside the text.
### Formula 
df["column"].str.strip()
### Related methods:
- .str.lstrip() removes characters from the left.
- .str.rstrip() removes characters from the right.
No mathematical formula is required.
### When it is used
- To remove extra spaces from text values.
- To clean data imported from CSV or Excel files.
- To avoid incorrect comparisons caused by leading or trailing whitespace.
### AI/ML example
In a customer churn dataset, "Active " and "Active" could be treated as different categories. Using strip() helps standardize these values before analysis.

```python
df["Name_Clean"] = df["Name"].str.strip()

print(df[["Name", "Name_Clean"]])
```

We used .str.strip() to remove the extra spaces surrounding the names in the original column. For example, "  RAVI " becomes "RAVI" while " ARUN " becomes "ARUN". Spaces between words are preserved, so a name such as "Ravi Kumar" remains unchanged internally. This cleaning step helps prevent duplicate categories and incorrect matching when working with text data.

## split()

The split() method divides a string into multiple parts using a specified separator. In pandas, .str.split() applies this operation to each string in a column and returns lists or separate columns when expand=True is used.
### Formula 
df["column"].str.split(separator)
To create separate columns:
df["column"].str.split(separator, expand=True)
No mathematical formula is required.
### When it is used
- To separate full names into first and last names.
- To extract information from delimited text.
- To divide structured text fields into separate features.
### AI/ML example
In an NLP project, splitting a sentence into individual words is a basic text-processing step used in some tokenization workflows.

```python
df = pd.DataFrame({
    "Full_Name": [
        "Ravi Kumar",
        "Priya Sharma",
        "Arun Reddy"
    ]
})

df[["First_Name", "Last_Name"]] = (
    df["Full_Name"].str.split(" ", n=1, expand=True)
)

print(df)
```

We created a DataFrame containing full names and used .str.split(" ", n=1, expand=True) to divide each name at the first space. The n=1 parameter limits the split to one occurrence, and expand=True returns the results as separate columns instead of lists. We assigned these columns to First_Name and Last_Name, making it easier to search, filter, or analyze each part of a name independently.

## contains()

The contains() method checks whether a string contains a specified substring or pattern. In pandas, .str.contains() returns a Boolean Series indicating which values match the search expression.
### Formula 
df["column"].str.contains("pattern", na=False)
- na=False treats missing values as non-matches.
- By default, case=True makes the search case-sensitive.
- regex=True by default interprets the pattern as a regular expression.
No mathematical formula is required.
### When it is used
- To find rows containing particular words or phrases.
- To filter records using text conditions.
- To identify keywords in customer feedback, emails, or descriptions.
### AI/ML example
In a spam detection project, .str.contains() can help identify messages containing suspicious words such as "prize" or "winner" during exploratory analysis.

```python
df = pd.DataFrame({
    "Email": [
        "ravi@gmail.com",
        "priya@yahoo.com",
        "arun@gmail.com",
        "sneha@outlook.com",
        "kiran@gmail.com"
    ]
})

gmail_users = df[
    df["Email"].str.contains("@gmail.com", regex=False, na=False)
]

print(gmail_users)
```

We used .str.contains() to identify email addresses containing the literal text "@gmail.com". Setting regex=False ensures that pandas treats the search pattern as ordinary text rather than a regular expression. We also used na=False so missing email values would not cause problems during filtering. Finally, Boolean indexing returned only the rows belonging to Gmail users.

## replace()

The replace() method substitutes specified text or patterns with new values. In pandas, .str.replace() can replace text within strings across an entire column, while Series.replace() can replace complete cell values.
### Formula 
df["column"].str.replace("old", "new", regex=False)

No mathematical formula is required.
### When it is used
- To correct spelling mistakes in text.
- To remove unwanted characters or symbols.
- To standardize text values from different sources.
### AI/ML example
In a product review dataset, you can remove unwanted symbols or replace inconsistent product labels before processing the text for sentiment analysis.

```python
df = pd.DataFrame({
    "Phone": [
        "+91-98765-43210",
        "+91-91234-56789",
        "+91-99887-76655"
    ]
})

df["Clean_Phone"] = df["Phone"].str.replace(
    "-",
    "",
    regex=False
)

print(df)
```

We created a DataFrame containing phone numbers with hyphens and used .str.replace() to remove those hyphens. The arguments "-" and "" specify the old text and its replacement, respectively. Setting regex=False tells pandas to treat the old text as a literal string. The cleaned values are stored in a new column, preserving the original phone numbers for reference. In real projects, phone numbers should also be validated according to the relevant country format.

## Definition
Regular expressions, commonly called regex, are patterns used to search, match, extract, or replace text based on defined rules. Python's re module and pandas string methods support regex for processing complex text patterns such as email formats, phone numbers, dates, and identifiers.

### When it is used
- To validate or identify text patterns.
- To extract phone numbers, dates, or IDs.
- To clean unstructured text and prepare it for analysis.
### AI/ML example
In an NLP project, regex can remove URLs, extract hashtags, or identify order IDs from customer reviews before the text is passed to a Machine Learning model.

```python
import pandas as pd

df = pd.DataFrame({
    "Message": [
        "Order ID: 12345",
        "Order ID: 67890",
        "No order available",
        "Order ID: 54321"
    ]
})

# Find messages containing a five-digit number
df["Has_Order_ID"] = df["Message"].str.contains(
    r"\d{5}",
    regex=True,
    na=False
)

# Extract the five-digit order ID
df["Order_ID"] = df["Message"].str.extract(
    r"(\d{5})"
)

print(df)
```

We created a DataFrame containing order-related messages and used the regex pattern \d{5} to identify sequences of exactly five consecutive digits. The .str.contains() method returns True for messages containing such a sequence and False otherwise. We then used .str.extract() with parentheses around the pattern to capture the matching digits and store them in the Order_ID column. The message without an order number receives a missing value (NaN). This technique is useful for extracting structured information from text, although additional validation may be needed to confirm that an extracted number is a valid order ID.

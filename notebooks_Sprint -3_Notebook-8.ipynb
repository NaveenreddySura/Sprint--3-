# Data Manipulation 
Data manipulation in Python means modifying, organizing, and transforming a dataset to make it suitable for analysis or Machine Learning. Using pandas, we can rename columns, remove unwanted records, create new columns, transform values, and handle duplicate records.

```python
%mamba install pandas
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran"],
    "Age": [25, 30, 28, 24, 35],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [30000, 45000, 40000, 28000, 60000]
}

df = pd.DataFrame(data)

print(df)
```

## Rename Columns

Renaming columns means changing the existing column names in a DataFrame without changing the underlying data. Pandas provides the rename() method to rename selected columns or all columns.
### Formula
df.rename(columns={"old_name": "new_name"})
No mathematical formula is required.
### When it is used
- To make column names more readable.
- To standardize column names across datasets.
- To match column names expected by a data pipeline or ML workflow.
### AI/ML example
A dataset contains a column named cust_age. You rename it to Customer_Age to make feature names consistent with the rest of your project.

```python
df_renamed = df.rename(columns={
    "Name": "Employee_Name",
    "Salary": "Monthly_Salary"
})

print(df_renamed.head())
```

We used the rename() method to change Name to Employee_Name and Salary to Monthly_Salary. The dictionary specifies the old column names as keys and their new names as values. The result is stored in df_renamed, so the original DataFrame remains unchanged. This helps maintain consistent naming conventions when preparing datasets for analysis or model training.

## Drop Rows

Dropping rows means removing selected records from a DataFrame based on their index labels or specified conditions. The drop() method can remove rows by index, while Boolean filtering can remove rows that meet particular conditions.
### Formula 
df.drop(index=[row_index])
No mathematical formula is required.
### When it is used
- To remove irrelevant or invalid records.
- To remove records identified as errors.
- To exclude observations that do not meet analysis requirements.
### AI/ML example
In a house-price dataset, you may remove records with impossible negative house areas after investigating and confirming that they are invalid.

```python
# Remove the row with index 2
df_dropped = df.drop(index=[2])

print(df_dropped)
```

In the first example, drop(index=[2]) removes the record whose index label is 2, which belongs to Arun. In the second example, Boolean filtering retains only employees aged 25 or older, removing Sneha because she is 24. Both examples create a new DataFrame without modifying the original df. In real projects, rows should be removed only after checking that they are genuinely unnecessary or invalid.

## Drop Columns

Dropping columns means removing selected features from a DataFrame while retaining the remaining columns. The drop() method uses axis=1 to remove columns rather than rows.
### Formula 
df.drop(columns=["column_name"])
No mathematical formula is required.
### When it is used
- To remove unnecessary columns.
- To eliminate redundant or irrelevant features.
- To exclude identifiers or data that should not be used as model inputs.
### AI/ML example
When predicting employee salaries, you might remove a record ID column if it has no predictive value and is not required for joining other data.

```python
df_dropped_columns = df.drop(columns=["Department"])

print(df_dropped_columns)
```

We used drop(columns=["Department"]) to remove the Department column from the DataFrame. The Name, Age, and Salary columns remain available for further analysis. The operation is stored in df_dropped_columns, so the original DataFrame remains unchanged. In Machine Learning, removing a column can simplify the feature set, but you should verify that the column does not contain useful predictive information before dropping it.

## Insert Columns

Inserting a column means adding a new column at a specific position in a DataFrame. Pandas provides the insert() method, which lets you specify the column position, name, and values.
### Formula 
df.insert(position, "column_name", values)
No mathematical formula is required.
### When it is used
- To add a new feature to a dataset.
- To insert a calculated or derived column at a specific position.
- To organize the column order for reporting or analysis.
### AI/ML example
In a customer dataset, you can insert a new feature such as Annual_Income calculated from monthly income.

```python
df_inserted = df.copy()

df_inserted.insert(
    2,
    "Annual_Salary",
    df_inserted["Salary"] * 12
)

print(df_inserted)
```

We first created a copy of the original DataFrame so that the original data would remain unchanged. Then we inserted a new column called Annual_Salary at position 2, using the monthly salary multiplied by 12. For example, Ravi's annual salary is calculated as ₹30,000 × 12 = ₹3,60,000. The new column becomes the third column in the DataFrame, which is useful when adding calculated features in a preferred order.

## assign()

The assign() method creates a new DataFrame with additional columns or updated values for existing columns. It is particularly useful for creating calculated columns while keeping the original DataFrame unchanged.
### Formula 
df.assign(new_column=calculation)
No mathematical formula is required by pandas; the calculation depends on your requirements.
### When it is used
- To create new calculated columns.
- To add features during data preprocessing.
- To chain several DataFrame transformations together.
### AI/ML example
In a sales prediction model, you can create a Profit feature by subtracting the cost from the selling price.

```python
df_assigned = df.assign(
    Annual_Salary=df["Salary"] * 12,
    Salary_Bonus=df["Salary"] * 0.10
)

print(df_assigned[["Name", "Salary", "Annual_Salary", "Salary_Bonus"]])
```

We used assign() to create two new columns: Annual_Salary and Salary_Bonus. The annual salary is calculated by multiplying the monthly salary by 12, while the bonus is calculated as 10% of the monthly salary. The resulting DataFrame contains the original columns and the newly calculated columns, while df remains unchanged. Unlike insert(), assign() does not let you specify the position of a new column directly.

## apply()

The apply() method applies a function to the values in a pandas Series or along the rows or columns of a DataFrame. It is useful for performing custom transformations when a simple built-in operation is not sufficient.
### Formula 
df["column"].apply(function)
For row-wise operations:
df.apply(function, axis=1)
- axis=0 applies a function to each column.
- axis=1 applies a function to each row.
### When it is used
- To apply custom logic to column values.
- To create labels or classifications from numerical data.
- To perform row-wise calculations using multiple columns.
### AI/ML example
In a customer segmentation project, you can use apply() to classify customers into salary groups such as Low, Medium, and High.

```python
def salary_category(salary):
    if salary < 35000:
        return "Low"
    elif salary <= 45000:
        return "Medium"
    else:
        return "High"

df_applied = df.copy()

df_applied["Salary_Category"] = (
    df_applied["Salary"].apply(salary_category)
)

print(df_applied[["Name", "Salary", "Salary_Category"]])
```

We defined a function named salary_category() that assigns a category based on an employee's salary. Salaries below ₹35,000 are classified as Low, salaries from ₹35,000 to ₹45,000 are classified as Medium, and salaries above ₹45,000 are classified as High. We then used apply() to execute the function for each value in the Salary column and store the result in Salary_Category. This is useful when creating features from business rules, although vectorized pandas operations are generally preferable for simple transformations on large datasets.

map()

The map() method transforms values in a pandas Series using a dictionary, another Series, or a function. It is commonly used to replace category labels with standardized values or numerical representations.
### Formula 
df["column"].map({"old_value": "new_value"})

No mathematical formula is required.
### When it is used
- To convert category labels into numerical values.
- To standardize values using a mapping dictionary.
- To simplify categorical feature preparation.
### AI/ML example
In a customer dataset, you can map Yes to 1 and No to 0 for a binary feature, provided that this encoding is appropriate for the model.

```python
df_mapped = df.copy()

df_mapped["Department_Code"] = (
    df_mapped["Department"].map({
        "IT": 1,
        "HR": 2,
        "Finance": 3
    })
)

print(df_mapped[["Name", "Department", "Department_Code"]])
```

We created a dictionary that maps IT to 1, HR to 2, and Finance to 3. The map() method looks up each department in this dictionary and returns the corresponding code in the new Department_Code column. This is a simple way to encode categories. However, these numerical codes may imply an order to some ML algorithms, so one-hot encoding is often more appropriate for nominal categories such as departments.

## replace()

The replace() method substitutes specified values in a Series or DataFrame with new values. It can replace individual values, multiple values, or values matching specified patterns, depending on how it is configured.
### Formula 
df["column"].replace(old_value, new_value)
No mathematical formula is required.
### When it is used
- To correct inconsistent values.
- To replace placeholders with missing values.
- To standardize labels across a dataset.
### AI/ML example
A dataset may use "N/A" or "Unknown" to represent missing ages. You can replace these placeholders with proper missing values before imputing or removing incomplete records.

```python
data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Department": ["IT", "Human Resources", "IT", "HR"]
}

df_replace = pd.DataFrame(data)

df_replace["Department"] = (
    df_replace["Department"].replace({
        "Human Resources": "HR"
    })
)

print(df_replace)
```

We created a DataFrame in which the HR department is represented using two different labels: Human Resources and HR. We used replace() with a dictionary to standardize Human Resources as HR. This ensures that equivalent categories are represented consistently, preventing them from being incorrectly treated as separate categories during analysis or Machine Learning preprocessing.

## Duplicate Handling

Duplicate handling means identifying and managing rows that repeat the same information or values across selected columns. Pandas provides duplicated() to identify duplicate rows and drop_duplicates() to remove them.

### Numerical example
Suppose a dataset contains 100 rows, of which 95 are unique.
\[
100-95=5
\]

Therefore, there are 5 duplicate rows under that definition of uniqueness.
### When it is used
- To identify repeated records in a dataset.
- To remove duplicate transactions or customer records when appropriate.
- To improve data quality before analysis and model training.
### AI/ML example
In a fraud-detection dataset, repeated copies of the same transaction can distort model evaluation or the apparent frequency of events. Identifying duplicates helps prevent such data-quality problems, although repeated transactions may also be legitimate and should be investigated before removal.

```python
data = {
    "Name": ["Ravi", "Priya", "Arun", "Ravi", "Kiran"],
    "Age": [25, 30, 28, 25, 35],
    "Salary": [30000, 45000, 40000, 30000, 60000]
}

df_duplicates = pd.DataFrame(data)

# Identify duplicate rows
print(df_duplicates.duplicated())

# Count duplicate rows
print("Duplicate count:", df_duplicates.duplicated().sum())

# Remove duplicate rows
df_unique = df_duplicates.drop_duplicates()

print(df_unique)
```

We created a DataFrame containing five rows, where Ravi's record appears twice with identical values. The duplicated() method marks the first occurrence as False and the repeated occurrence as True, while sum() counts the duplicate rows. The drop_duplicates() method then removes the repeated record and retains the first occurrence by default. In real datasets, you can use subset=["Name"] or another appropriate set of columns to detect duplicates according to a business-specific definition rather than comparing every column.

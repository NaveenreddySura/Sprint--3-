# Notebook 11 – Date & Time Operations 
Date and time operations in Python involve working with dates, times, durations, and time zones. In pandas, these operations help us analyze time-based data, calculate time differences, extract useful date features, and prepare datasets for Data Analysis and Machine Learning.

## Datetime

Datetime represents a specific date and time, including the year, month, day, hour, minute, and second. Python provides the datetime module, while pandas provides datetime64 data types for working with date and time values across a dataset.
### Formula 
pd.to_datetime(values)
No mathematical formula is required.
### When it is used
- To store and process dates and times consistently.
- To sort records chronologically.
- To perform date-based analysis and calculations.
### AI/ML example
In a sales forecasting model, datetime values help identify when each sale occurred and allow the model to analyze daily, monthly, or seasonal sales patterns.

```python
%mamba install pandas
```

```python
import pandas as pd

df = pd.DataFrame({
    "Event": ["Login", "Purchase", "Logout"],
    "Date_Time": [
        "2026-10-10 09:30:00",
        "2026-10-10 10:15:00",
        "2026-10-10 11:00:00"
    ]
})

df["Date_Time"] = pd.to_datetime(df["Date_Time"])

print(df)
print(df.dtypes)
```

We created a DataFrame containing three user events and their timestamps. Initially, the dates were stored as strings, so we used pd.to_datetime() to convert the Date_Time column into a proper pandas datetime type. The dtypes output confirms that pandas now recognizes the column as datetime data, allowing us to sort events by time, calculate durations, and extract date-related information.

## Timestamp

A timestamp represents a specific point in time, commonly expressed as a date and time or as a numeric count from a reference point. In pandas, Timestamp represents an individual date-time value and supports operations such as comparison, formatting, and date extraction.
### Formula
Unix timestamp conversion uses a reference point called the Unix epoch: 1970-01-01 00:00:00 UTC.
\[
\text{Unix Timestamp}=\text{Elapsed Seconds Since Epoch}
\]

### Numerical example
If a timestamp represents 60 seconds after the Unix epoch, it corresponds to:
1970-01-01 00:01:00 UTC
### When it is used
- To represent the exact time an event occurred.
- To compare event times or record creation times.
- To work with logs, transaction records, and sensor data.
### AI/ML example
In fraud detection, timestamps help establish the order of transactions and identify multiple suspicious transactions occurring within a short period.

```python
import pandas as pd

timestamp = pd.Timestamp("2026-10-16 10:30:00")

print(timestamp)
print("Year:", timestamp.year)
print("Month:", timestamp.month)
print("Day:", timestamp.day)
print("Hour:", timestamp.hour)
```

We created an individual timestamp using pd.Timestamp() to represent October 16, 2026, at 10:30 AM. We then accessed its year, month, day, and hour using the corresponding attributes. Unlike a complete DataFrame datetime column, which can contain many timestamps, a Timestamp object represents one specific date and time. This is useful when working with individual event times or comparing dates.

## Date Parsing

Date parsing means converting date and time values stored as strings or other formats into proper datetime objects. Pandas uses pd.to_datetime() to recognize dates, and you can specify a format when the input follows a known pattern.
### Formula 
pd.to_datetime(values, format="%d-%m-%Y")
### Common format codes:
- %Y — four-digit year
- %m — month number
- %d — day
- %H — hour in 24-hour format
- %M — minute
- %S — second
### When it is used
- To convert date strings into usable datetime values.
- To handle datasets containing dates in different formats.
- To prevent incorrect date interpretation during analysis.
### AI/ML example
In a demand forecasting project, transaction dates may arrive as strings such as "16-10-2026". Parsing them correctly allows the model pipeline to extract the correct month, day, and weekday.

```python
import pandas as pd

df = pd.DataFrame({
    "Order_ID": [101, 102, 103],
    "Order_Date": [
        "16-10-2026",
        "17-10-2026",
        "18-10-2026"
    ]
})

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    format="%d-%m-%Y"
)

print(df)
print(df.dtypes)
```

We created a DataFrame with order dates stored as strings in day-month-year format. The pd.to_datetime() function converts those strings into proper datetime values, and the format="%d-%m-%Y" argument tells pandas exactly how to interpret the dates. This is particularly important when a date such as 04-05-2026 could mean either April 5 or May 4 depending on the source format. Correct parsing ensures that subsequent date calculations and analyses use the intended dates.

## Timedelta

A Timedelta represents a duration or difference between two points in time. Pandas supports adding or subtracting timedeltas from dates and calculating the time elapsed between events.
### Formula
\[
\text{Duration}=\text{End Time}-\text{Start Time}
\]

### Numerical example
Suppose an order is placed at 10:00 AM and delivered at 1:30 PM.
\[
\text{Duration}=3\text{ hours }30\text{ minutes}
\]

### When it is used
- To calculate the duration between events.
- To find delivery times or session lengths.
- To add or subtract time from a date.
### AI/ML example
In a delivery prediction model, timedelta can calculate how long each delivery took. These durations can then be used to identify patterns and train a model to estimate future delivery times.

```python
import pandas as pd

start_time = pd.Timestamp("2026-10-16 10:00:00")
end_time = pd.Timestamp("2026-10-16 13:30:00")

duration = end_time - start_time

print("Duration:", duration)
print("Hours:", duration.total_seconds() / 3600)

# Add 2 days to the starting time
new_time = start_time + pd.Timedelta(days=2)

print("New time:", new_time)
```

We created two timestamps representing the start and end of an event and subtracted the start time from the end time to calculate a duration of three hours and thirty minutes. We then converted that duration into hours by dividing the total seconds by 3,600. Finally, we added a two-day Timedelta to the starting timestamp to calculate a future time. These operations are useful for measuring durations, calculating deadlines, and creating time-based features for Machine Learning.

## Time Zones

A time zone identifies the local time rules used in a particular geographical region. Pandas supports time-zone-aware timestamps, allowing you to localize timestamps to a time zone and convert them between time zones while accounting for applicable offset rules.
### Formula
No mathematical formula is required. Time-zone conversion depends on the UTC offset applicable at the specified date and location.
### When it is used
- To combine event data from different countries.
- To convert UTC timestamps into local time.
- To avoid errors when comparing timestamps from different time zones.
### AI/ML example
In a global e-commerce dataset, purchases may occur in India, the United States, and the United Kingdom. Converting event times to a consistent reference such as UTC makes it easier to analyze transaction order and build time-based features.

```python
import pandas as pd

# Create a timezone-aware timestamp in India
india_time = pd.Timestamp(
    "2026-10-16 10:00:00",
    tz="Asia/Kolkata"
)

# Convert the same instant to UTC
utc_time = india_time.tz_convert("UTC")

print("India time:", india_time)
print("UTC time:", utc_time)
```

We created a time-zone-aware timestamp representing 10:00 AM in India using the Asia/Kolkata time zone. We then used tz_convert("UTC") to represent the same instant in Coordinated Universal Time, which is 4:30 AM UTC. The actual instant does not change; only its displayed local time and offset change. This distinction is essential when combining time-based records from multiple countries.
Important: Use tz_localize() when assigning a time zone to a timestamp that currently has no time-zone information. Use tz_convert() when converting an already time-zone-aware timestamp to another time zone.

## Extracting Date Features

Extracting date features means deriving useful components such as year, month, day, weekday, hour, and quarter from a datetime column. These components can help identify seasonal patterns, business cycles, and time-related trends in a dataset.
### Formula
No single mathematical formula is required because each feature is extracted directly from the date or timestamp.
### When it is used
- To analyze monthly, weekly, or seasonal trends.
- To identify weekday versus weekend behavior.
- To prepare time-based input features for Machine Learning models.
### AI/ML example
In a retail sales forecasting model, the month, day of the week, and holiday period can help explain changes in demand. For example, sales may be higher on weekends or during particular months.

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": [
        "2026-10-16 10:30:00",
        "2026-11-21 15:45:00",
        "2026-12-25 12:00:00"
    ],
    "Sales": [5000, 7500, 12000]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Weekday"] = df["Order_Date"].dt.day_name()
df["Hour"] = df["Order_Date"].dt.hour
df["Quarter"] = df["Order_Date"].dt.quarter

print(df)
```

We created a sales dataset containing order dates and sales amounts, then converted the Order_Date column into datetime format. Using the .dt accessor, we extracted the year, month, day, weekday name, hour, and quarter into separate columns. For example, the order dated December 25, 2026, is identified as occurring in December, on a Friday, at 12:00 PM, during the fourth quarter. These features can help a Machine Learning model discover recurring sales patterns and relationships between dates and sales.

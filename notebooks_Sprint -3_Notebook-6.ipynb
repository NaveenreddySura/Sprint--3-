# Data Inspection 
Data inspection in Python is the process of examining a dataset to understand its structure, columns, data types, missing values, and value distribution before performing data analysis or building Machine Learning models.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arjun", "Sneha", "Kiran", "Ravi"],
    "Age": [25, 28, 23, 30, 27, 25],
    "Salary": [30000, 45000, 25000, 60000, 40000, 30000],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR", "IT"
    ],
    "Experience": [2, 5, 1, 7, 4, 2]
}

df = pd.DataFrame(data)

print(df)
```

## head()

The head() method displays the first few rows of a DataFrame. By default, it returns the first five rows, but you can specify the number of rows you want.

### When is it used?
- To quickly preview a newly loaded dataset.
- To check whether data was imported correctly.
- To examine column values before cleaning or analysis.
### AI/ML example
When building a salary prediction model, you can use head() to inspect employee ages, salaries, and experience before selecting features for training.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Manoj"],
    "Age": [25, 30, 28, 24, 35, 27],
    "Salary": [30000, 45000, 40000, 28000, 60000, 35000]
}

df = pd.DataFrame(data)

print(df.head())
```

We created a DataFrame containing employee names, ages, and salaries. The df.head() method displays the first five rows by default, so Manoj's record is not displayed because it is the sixth row. If we use df.head(3), Python displays only the first three rows.

## tail()

The tail() method displays the last few rows of a DataFrame. By default, it returns the last 5 rows, and you can specify a different number when required.

### When it is used
- To inspect the last records in a dataset.
- To check whether data was appended or loaded correctly.
- To examine the most recent records in time-ordered data.
### AI/ML example
When training a model using historical sales data, tail() can help you inspect the latest sales records and confirm that the newest dates were loaded.

```python
print(df.tail(3))
```

The df.tail(3) method selects the final three rows of the DataFrame. It displays Sneha, Kiran, and Manoj because their records are the last three entries. Unlike head(), which starts at the beginning, tail() starts from the end of the DataFrame.

## sample()

The sample() method selects a random sample of rows from a DataFrame. You can choose how many rows to select and use a random seed to make the selection reproducible.

### When it is used
- To inspect randomly selected records.
- To perform quick exploratory data analysis.
- To check a representative subset of a large dataset.
### AI/ML example
If a dataset contains one million customer records, you can randomly inspect 100 records before beginning feature engineering.

```python
print(df.sample(n=3, random_state=42))
```

The df.sample() method selects three rows randomly rather than choosing rows from the beginning or end. We set random_state=42 so that the selection remains the same when the same DataFrame and sampling settings are used. This is useful when inspecting large datasets without printing every record.

## info()

The info() method provides a concise summary of a DataFrame, including its column names, number of non-null values, data types, and memory usage. It is especially useful for checking missing data and understanding the structure of a dataset.

### When it is used
- To identify columns containing missing values.
- To check whether columns have appropriate data types.
- To inspect the number of rows and approximate memory usage.
### AI/ML example
Before training a customer churn model, info() helps identify missing customer ages and columns stored with incorrect data types.

```python
data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Age": [25, 30, None, 24],
    "Salary": [30000, 45000, 40000, 28000]
}

df = pd.DataFrame(data)

df.info()
```

We created a DataFrame containing four employees, with a missing value in the Age column. The info() method shows that Age has only three non-null values, which indicates that one age is missing. It also shows each column's data type, helping us determine whether data cleaning or type conversion is necessary before further analysis.

## describe()

The describe() method generates descriptive statistics that summarize the distribution of numerical columns in a DataFrame. By default, it displays count, mean, standard deviation, minimum, quartiles, and maximum.
### Formula
For numerical data, the mean is calculated as:
\[
\text{Mean}=\frac{\sum x_i}{n}
\]

The standard deviation measures how spread out the values are around the mean.
### Numerical example
Consider the salaries:
20000, 30000, 40000
The mean salary is:
\[
\frac{20000+30000+40000}{3}
=30000
\]

Therefore, the average salary is ₹30,000.
### When it is used
- To understand the distribution of numerical data.
- To identify unusually high or low values.
- To compare the central tendency and spread of features.
### AI/ML example
For a house-price prediction model, describe() helps examine the average house price, the range of house sizes, and the variation in prices.

```python
df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [20000, 30000, 40000, 50000, 60000]
})

print(df.describe())
```

We created a DataFrame with five age and salary values and used describe() to calculate summary statistics for both columns. The output shows an average age of 40 and an average salary of ₹40,000, along with the minimum, maximum, and quartile values. These statistics help us understand the dataset before selecting features or training a model.

## shape

The shape attribute returns the dimensions of a DataFrame as a tuple containing the number of rows and columns. It is an attribute rather than a method, so you do not use parentheses.

### When it is used
- To determine the size of a dataset.
- To verify the number of records and features.
- To check how the dataset changes after filtering or removing columns.
### AI/ML example
Before training a model, shape can confirm that a dataset has 10,000 observations and 15 features.

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 30, 28],
    "Salary": [30000, 45000, 40000]
})

print(df.shape)
```

We created a DataFrame containing three employee records and three columns. The df.shape attribute returns (3, 3), meaning the dataset has three rows and three columns. This is useful when checking the dimensions of a dataset before and after data preprocessing.

## columns

The columns attribute returns the names of all columns in a DataFrame. It allows you to inspect column names and use them when selecting, renaming, or processing features.

### When it is used
- To view all feature names.
- To check column names before selecting specific features.
- To rename columns or verify expected fields.
### AI/ML example
In a customer churn model, you can inspect columns such as Age, MonthlyCharges, and ContractType before choosing model features.

The df.columns attribute returns the names of the three columns in the DataFrame as a pandas Index object. This lets us confirm which fields are available in the dataset. For example, we could later select only the Age and Salary columns for numerical analysis.

## dtypes

The dtypes attribute displays the data type of each DataFrame column. It helps identify whether values are stored as integers, floating-point numbers, text, dates, or other supported types.

### When it is used
- To identify incorrect data types.
- To decide whether type conversion is necessary.
- To prepare features for data analysis and machine learning.
### AI/ML example
If a dataset stores house prices as text instead of numbers, dtypes helps identify the issue before calculating statistics or training a regression model.

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 30, 28],
    "Salary": [30000.5, 45000.0, 40000.0]
})

print(df.dtypes)
```

We created a DataFrame containing a text column, an integer column, and a floating-point column. The dtypes attribute shows that Name is stored as object, Age as an integer, and Salary as a floating-point number. Understanding these types helps us identify columns that might need conversion before performing calculations or training a model.

## memory_usage()

The memory_usage() method reports how much memory each DataFrame column uses to store its data. By default, it reports the memory used by the column values, and the deep=True option can provide a more detailed estimate for object-based data.
### Formula
No fixed mathematical formula is required for everyday use.
Memory usage is generally reported in bytes.
\[
1\text{ KB}=1024\text{ bytes}
\]

### Numerical example
If a column uses 2,048 bytes, its memory usage is:
\[
\frac{2048}{1024}=2\text{ KB}
\]

### When it is used
- To inspect the memory footprint of a DataFrame.
- To optimize memory usage when working with large datasets.
- To compare memory use before and after changing data types.
### AI/ML example
When processing millions of transaction records, memory_usage() helps identify memory-intensive columns so you can consider suitable data types or other optimizations.

```python
df = pd.DataFrame({
    "Age": [25, 30, 28],
    "Salary": [30000, 45000, 40000]
})

print(df.memory_usage())
print(df.memory_usage(deep=True))
```

We created a small DataFrame with two numerical columns and used memory_usage() to estimate the memory occupied by each column. The output includes memory used by the index as well as the individual columns. When deep=True is used, pandas makes a deeper estimate for object-based data, which is particularly helpful when assessing memory usage in larger datasets.

## nunique()

The nunique() method counts the number of distinct values in a Series or in each DataFrame column. By default, it excludes missing values from the count.
### Formula
No mathematical formula is required; the method counts the unique values.
### Numerical example
Consider the values:
[10, 20, 20, 30, 30]
The distinct values are 10, 20, and 30.
Therefore:
\[
\text{Number of unique values}=3
\]

### When it is used
- To identify the number of distinct categories in a column.
- To detect columns containing repeated values.
- To assess whether a feature has too many unique values for a particular encoding strategy.
### AI/ML example
In a customer dataset, nunique() can reveal how many unique cities or customer IDs exist. This helps distinguish categorical features from identifier columns.

```python
df = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Hyderabad", "Bengaluru"],
    "Age": [25, 30, 25, 35]
})

print(df.nunique())
print("Unique cities:", df["City"].nunique())
```

We created a DataFrame with city and age columns containing repeated values. The df.nunique() method counts the distinct values in each column, showing three unique cities and three unique ages. We then used df["City"].nunique() to count distinct city names specifically. This is useful for understanding categorical features before preparing data for machine learning.

## value_counts()

The value_counts() method counts how many times each distinct value appears in a pandas Series. By default, it sorts the results from the most frequent value to the least frequent value and excludes missing values.
### Formula
The frequency of a value is the number of times that value occurs in the dataset.
\[
\text{Frequency}(x)=\text{Count of occurrences of }x
\]

### Numerical example
Consider:
["Pass", "Fail", "Pass", "Pass", "Fail"]
- Pass appears 3 times.
- Fail appears 2 times.
The total number of observations is \(3+2=5\).
### When it is used
- To understand the frequency of categories.
- To detect class imbalance in classification datasets.
- To inspect the distribution of labels or categorical features.
### AI/ML example
In a fraud-detection dataset, value_counts() can show that most transactions are legitimate and only a small number are fraudulent. This helps identify class imbalance before model training.

```python
df = pd.DataFrame({
    "Result": ["Pass", "Fail", "Pass", "Pass", "Fail", "Pass"]
})

print(df["Result"].value_counts())
```

We created a DataFrame containing the results of six students and used value_counts() on the Result column. Python counts four occurrences of Pass and two occurrences of Fail, then displays the most frequent category first. This helps us understand category distributions and is especially useful for checking whether target classes in a classification problem are balanced.

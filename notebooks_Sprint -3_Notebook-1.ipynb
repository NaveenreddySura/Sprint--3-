# NumPy & Pandas for AI/ML Engineers

## What Is NumPy?

NumPy stands for Numerical Python. It is a Python library that provides efficient multidimensional arrays and mathematical functions for performing numerical computations on large datasets.
It forms the foundation of many data science and machine learning libraries, including Pandas, SciPy, and Scikit-learn.
### Formula
There is no specific formula for NumPy because it is a library rather than a mathematical operation.
### Numerical Example
Suppose we have the following marks:

Mathematics: 80
Science: 90
English: 85
Adding 5 bonus marks to every subject gives:

[80, 90, 85] + 5 = [85, 95, 90]

### Where It Is Used
- Numerical and statistical calculations.
- Scientific computing.
- Data processing and analysis.
- Machine learning and deep learning.
### AI/ML Engineer Example
An AI/ML engineer uses NumPy to store customer data as numerical arrays, calculate statistics, and prepare data for a machine learning model.

```python
import numpy as np

scores = np.array([70, 80, 90, 85, 75])

print("Scores:", scores)
print("Average:", np.mean(scores))
print("Maximum:", np.max(scores))
```

We imported NumPy using the alias np and created an array containing five exam scores with np.array(). We used np.mean() to calculate the average score and np.max() to find the highest score. NumPy performs these calculations directly on the array without requiring us to write separate loops.

## Why NumPy?

NumPy is useful because its arrays generally require less memory and support faster numerical operations than equivalent Python lists for large numerical datasets. It also provides vectorized operations, multidimensional arrays, broadcasting, and mathematical functions that simplify data processing.

Unlike a normal Python list, a NumPy array allows us to apply many operations to all its elements directly.

### Formula
For increasing each value by a fixed amount:

[ y_i = x_i + c ]

Where:

(x_i) is an original value.
(c) is the added amount.
(y_i) is the updated value.
###  Numerical Example
Original values: [10, 20, 30]

Add 5 to each value: [15, 25, 35]
### Where It Is Used
- Processing large numerical datasets.
- Performing calculations without explicit Python loops.
- Preparing arrays for machine learning algorithms.
- Applying mathematical operations to entire datasets.
### AI/ML Engineer Example
An AI/ML engineer processes thousands of customer purchase amounts. Instead of calculating a tax amount for every customer using a manual loop, the engineer can multiply the entire NumPy array by the tax rate in one operation.

```python
import numpy as np

prices = np.array([100, 200, 300, 400])

tax = prices * 0.10

print("Prices:", prices)
print("Tax:", tax)
print("Final prices:", prices + tax)
```

We created a NumPy array containing four product prices. We multiplied the entire array by 0.10 to calculate a 10% tax for each product. Then, we added the tax array to the original prices to calculate the final prices. NumPy automatically applies the operation to every element, which is called vectorization.

## NumPy Arrays (ndarray)

An ndarray (N-dimensional array) is NumPy's main data structure for storing elements of the same data type in one or more dimensions. It supports efficient indexing, mathematical operations, and manipulation of numerical data.
### Formula
For a one-dimensional array containing (n) elements:

[ A = [a_1, a_2, \ldots, a_n] ]

### Numerical Example
An array with 2 rows and 3 columns contains:
\[
2\times3=6\text{ elements}
\]

### Where It Is Used
- Storing numerical datasets.
- Representing tables and matrices.
- Preparing data for ML algorithms.
- Performing mathematical operations on multiple values.
### AI/ML Engineer Example
An AI/ML engineer represents customer information, such as age, income, and spending score, as a NumPy array before passing it to a machine learning model.

```python
import numpy as np

arr = np.array([
    [25, 30000, 70],
    [30, 40000, 80]
])

print(arr)
print("Type:", type(arr))
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
```

We created a two-dimensional NumPy array containing two customers and three numerical features for each customer. The type() function confirms that the object is an ndarray. The ndim attribute returns the number of dimensions, shape returns the number of rows and columns, and size returns the total number of elements.

## Creating Arrays

Creating arrays means converting values into NumPy arrays or generating arrays using predefined patterns. NumPy provides functions such as array(), zeros(), ones(), arange(), and linspace() to create arrays for different numerical tasks.
### Formula
For an evenly spaced sequence with step size \(s\):
\[
a_n=a_0+ns
\]

### Numerical Example
Starting at 2 with a step size of 2 gives:
[2, 4, 6, 8, 10]
Where It Is Used
- Creating datasets for analysis.
- Initializing model parameters.
- Generating numerical sequences.
- Creating arrays filled with zeros or ones.
### AI/ML Engineer Example
An AI/ML engineer creates an array of evenly spaced values to test how a model behaves across different input values.

```python
import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.zeros(4)
arr3 = np.ones(3)
arr4 = np.arange(2, 12, 2)
arr5 = np.linspace(0, 1, 5)

print("Array:", arr1)
print("Zeros:", arr2)
print("Ones:", arr3)
print("Range:", arr4)
print("Linspace:", arr5)
```

We used five different NumPy functions to create arrays. array() converts a list into an array, zeros() creates an array of zeros, and ones() creates an array of ones. arange() generates values using a specified step size, while linspace() generates a specified number of evenly spaced values between two endpoints, including both endpoints by default.

## Dimensions

Dimensions describe the number of axes in a NumPy array. A one-dimensional array contains a single axis, a two-dimensional array has rows and columns, and a three-dimensional array can represent multiple two-dimensional tables stacked together.
### Formula
The number of dimensions is given by:
\[
\text{Dimensions}=\texttt{array.ndim}
\]

### Numerical Example
- [10, 20, 30] → 1 dimension.
- [[10, 20], [30, 40]] → 2 dimensions.
- A stack of matrices → 3 dimensions.
### Where It Is Used
- Representing tables and matrices.
- Working with image and video data.
- Preparing inputs for machine learning and deep learning.
### AI/ML Engineer Example
An AI/ML engineer represents a grayscale image as a two-dimensional array of pixel values and a batch of images as a three-dimensional array.

```python
import numpy as np

arr1 = np.array([10, 20, 30])

arr2 = np.array([
    [10, 20],
    [30, 40]
])

arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D dimensions:", arr1.ndim)
print("2D dimensions:", arr2.ndim)
print("3D dimensions:", arr3.ndim)
```

We created three arrays with different structures: a simple list of values, a matrix, and a stack of matrices. We used the ndim attribute to determine the number of axes in each array. The results show how NumPy represents data with increasing structural complexity.

## Shape

The shape of a NumPy array describes how many elements it contains along each dimension. For a two-dimensional array, the shape is represented as (rows, columns), while higher-dimensional arrays contain one size value for every axis.
### Formula
For a two-dimensional array:
\[
\text{Shape}=(\text{Rows},\text{Columns})
\]

### Numerical Example
An array containing 3 rows and 2 columns has the shape (3, 2).
### Where It Is Used
- Checking dataset structure.
- Verifying input dimensions before model training.
- Reshaping arrays.
- Debugging incompatible array operations.
### AI/ML Engineer Example
An AI/ML engineer checks whether a dataset contains 1,000 observations and 5 features before sending it to a machine learning model.

```python
import numpy as np

data = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])

print("Shape:", data.shape)
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
```

We created a two-dimensional array with three rows and two columns. The shape attribute returned (3, 2). We accessed shape[0] to get the number of rows and shape[1] to get the number of columns. This is useful for checking whether the dimensions of data match the requirements of a machine learning algorithm.

## Size

The size of a NumPy array is the total number of elements it contains, regardless of how many dimensions the array has. It is different from shape, which describes the number of elements along each axis.
### Formula
\[
\text{Size}=\prod_{i=1}^{n}d_i
\]

### Numerical Example
An array with shape (3, 4) contains:
\[
3\times4=12\text{ elements}
\]

### Where It Is Used
- Counting elements in an array.
- Checking dataset dimensions.
- Calculating memory requirements.
- Verifying reshaping operations.
### AI/ML Engineer Example
An AI/ML engineer checks the number of pixel values in an image array before processing it in an image-classification model.

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Shape:", arr.shape)
print("Size:", arr.size)
```

We created a two-dimensional array containing two rows and three columns. The shape attribute returned (2, 3), while the size attribute returned 6 because the array contains six elements in total.

## Data Types (dtype)

The data type (dtype) specifies the kind of values stored in a NumPy array, such as integers, floating-point numbers, or Boolean values. NumPy arrays generally use a single data type for their elements, which helps improve memory efficiency and numerical processing.
### Formula
No mathematical formula is required. Use array.dtype to inspect an array's data type.
Numerical Example
- [10, 20, 30] → integer array.
- [10.5, 20.5, 30.5] → floating-point array.
### Where It Is Used
- Controlling memory usage.
- Ensuring numerical calculations use appropriate types.
- Preparing data for machine learning algorithms.
- Converting data between integer and floating-point formats.
### AI/ML Engineer Example
An AI/ML engineer converts image pixel values to floating-point numbers before normalizing them to values between 0 and 1.

```python
import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([10.5, 20.5, 30.5])
arr3 = np.array([1, 2, 3], dtype=float)

print("Integer dtype:", arr1.dtype)
print("Float dtype:", arr2.dtype)
print("Converted array:", arr3)
print("Converted dtype:", arr3.dtype)
```

We created an integer array and a floating-point array, then inspected their data types using dtype. We also passed dtype=float to np.array() to convert integer values into floating-point values. The numerical values remain the same, but their representation changes.

## Indexing

Indexing is the process of accessing individual elements of a NumPy array using their positions. NumPy uses zero-based indexing, meaning the first element has index 0, the second has index 1, and negative indices can access elements from the end.
### Formula
For a one-dimensional array, the element at position \(i\) is accessed using:
array[i]
### Numerical Example
For [10, 20, 30, 40, 50]:
- Index 0 → 10
- Index 2 → 30
- Index -1 → 50
### Where It Is Used
- Retrieving specific values.
- Accessing rows and columns.
- Extracting individual observations.
- Updating array elements.
### AI/ML Engineer Example
An AI/ML engineer accesses a particular feature value or a specific customer's record when debugging a dataset.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])

arr[1] = 25

print("Updated array:", arr)
```

We created a one-dimensional array and accessed its first, third, and last elements using positive and negative indices. Then, we changed the element at index 1 from 20 to 25. NumPy allows direct access and modification of individual array elements through indexing.

## Slicing

Slicing is used to extract a section of an array by specifying a start position, a stop position, and an optional step size. Unlike indexing, which normally returns a single element, slicing returns a sequence of elements.

### Formula
array[start:stop:step]

### Numerical Example
For [10, 20, 30, 40, 50, 60]:
arr[1:4] returns [20, 30, 40].
### Where It Is Used
- Extracting a subset of a dataset.
- Selecting specific rows or columns.
- Processing parts of an array.
- Preparing training and testing data.
### AI/ML Engineer Example
An AI/ML engineer selects a subset of rows from a large dataset to test a preprocessing function before applying it to the full dataset.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

print("Elements 1 to 3:", arr[1:4])
print("First three:", arr[:3])
print("Last three:", arr[-3:])
print("Every second element:", arr[::2])
```

We created an array containing six values and used different slicing expressions to select portions of it. arr[1:4] selects elements from index 1 up to, but not including, index 4. arr[:3] selects the first three values, arr[-3:] selects the last three, and arr[::2] selects every second element.

## Reshaping

Reshaping changes the dimensions of a NumPy array without changing the total number or order of its elements. The new shape must be compatible with the array's total size.
### Formula
\[
\text{Old Size}=\text{New Size}
\]

For example:
\[
2\times3=6=3\times2
\]

### Numerical Example
The array [1, 2, 3, 4, 5, 6] can be reshaped from (6,) into (2, 3).
### Where It Is Used
- Preparing data for machine learning models.
- Converting a flat array into a matrix.
- Restructuring image data.
- Changing the arrangement of data for mathematical operations.
### AI/ML Engineer Example
An AI/ML engineer reshapes a one-dimensional array of six measurements into a matrix with two observations and three features.

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

reshaped = arr.reshape(2, 3)

print("Original shape:", arr.shape)
print("Reshaped array:")
print(reshaped)
print("New shape:", reshaped.shape)
```

We created a one-dimensional array containing six elements and used reshape(2, 3) to arrange them into two rows and three columns. We then printed the original and new shapes to show that the number of dimensions changed while the total number of elements remained six.

## Flatten

Flattening converts a multidimensional array into a one-dimensional array containing all its elements in sequence. It is useful when a function or algorithm requires a single sequence rather than a matrix or higher-dimensional structure.
### Formula
For an array containing \(n\) total elements, flattening produces a one-dimensional array of shape (n,).
### Numerical Example
Input:
[[1, 2, 3],
 [4, 5, 6]]


Flattened result:
[1, 2, 3, 4, 5, 6]
### Where It Is Used
- Converting matrices into vectors.
- Preparing image data for certain ML algorithms.
- Simplifying multidimensional data.
- Passing data to functions that require one-dimensional input.
### AI/ML Engineer Example
An AI/ML engineer converts a small matrix of image pixel values into a one-dimensional array before passing it to a model that expects flattened input.

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = arr.flatten()

print("Original array:")
print(arr)

print("Flattened array:", flat)
```

We created a two-dimensional array and used flatten() to convert it into a one-dimensional array. The output contains the same six elements in row-major order, but the original two-dimensional arrangement is no longer present in the flattened result. Unlike many basic slicing operations, flatten() returns a copy of the data.

## Copy vs View
A Copy is a separate array with its own data, so modifying it does not change the original array. A View is a different array object that shares the same underlying data as the original, so modifying a view can also modify the original array.
### Formula
No mathematical formula is required. The key distinction is whether the arrays share the same underlying data.
### Numerical Example
Original array: [10, 20, 30]
- Copy changed to [99, 20, 30] → original remains [10, 20, 30].
- View changed to [99, 20, 30] → original can also become [99, 20, 30].
### Where It Is Used
- Avoiding unintended changes to datasets.
- Reducing unnecessary memory usage.
- Creating independent datasets for preprocessing.
- Working efficiently with array subsets.
### AI/ML Engineer Example
An AI/ML engineer creates a copy of a dataset before applying transformations so that the original data remains unchanged. A view may be used when the engineer intentionally wants to modify a portion of the original array.

```python
import numpy as np

original = np.array([10, 20, 30])

copy_arr = original.copy()
view_arr = original.view()

copy_arr[0] = 99

print("After changing copy:")
print("Original:", original)
print("Copy:", copy_arr)

view_arr[1] = 88

print("\nAfter changing view:")
print("Original:", original)
print("View:", view_arr)
```

We created an original array and generated a separate copy using copy() and a shared-data view using view(). When we changed the first element of the copy, the original remained unchanged. When we changed the second element of the view, the original also changed because both arrays share the same underlying data.


## Broadcasting
Definition
Broadcasting is a NumPy mechanism that allows arrays with compatible shapes to participate in arithmetic operations without manually repeating their values. NumPy compares dimensions from right to left and allows dimensions to match or for one of them to equal 1.
### Formula
Two dimensions are compatible when:
\[
a=b\quad\text{or}\quad a=1\quad\text{or}\quad b=1
\]

For example, a matrix with shape (2, 3) can be added to an array with shape (3,).
### Numerical Example
Matrix:
\[
A=\begin{bmatrix}1&2&3\\4&5&6\end{bmatrix}
\]

Array:
\[
B=[10,20,30]
\]

The result is:
\[
A+B=\begin{bmatrix}11&22&33\\14&25&36\end{bmatrix}
\]

### Where It Is Used
- Adding values to rows or columns of a matrix.
- Normalizing numerical datasets.
- Performing vectorized calculations.
- Applying feature-specific transformations.
### AI/ML Engineer Example
An AI/ML engineer subtracts the mean of each feature from every observation in a dataset. Broadcasting allows the mean array to be applied to all rows without manually copying it.

```python
import numpy as np

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

values = np.array([10, 20, 30])

result = matrix + values

print(result)
```

We created a two-dimensional matrix with two rows and three columns and a one-dimensional array containing three values. We added the arrays directly using matrix + values. NumPy broadcast the one-dimensional array across both rows, adding 10 to the first column, 20 to the second, and 30 to the third. No manual repetition was necessary.

## Array Operations

Array Operations are mathematical, comparison, and logical operations performed on NumPy arrays. These operations generally work element by element, allowing calculations on entire arrays without writing explicit loops.
### Formula
For element-wise multiplication:
\[
C_i=A_i\times B_i
\]

For element-wise addition:
\[
C_i=A_i+B_i
\]

### Numerical Example
If:
- \(A=[2,4,6]\)
- \(B=[1,2,3]\)
Then:
- Addition: [3, 6, 9]
- Multiplication: [2, 8, 18]
### Where It Is Used
- Calculating totals and averages.
- Applying mathematical transformations.
- Comparing numerical values.
- Processing large datasets efficiently.
### AI/ML Engineer Example
An AI/ML engineer calculates the total revenue for multiple products by multiplying each product's selling price by its quantity sold.

```python
import numpy as np

prices = np.array([100, 200, 300])
quantity = np.array([2, 3, 4])

total = prices * quantity
discounted = prices - 10
squared = prices ** 2

print("Total revenue:", total)
print("Discounted prices:", discounted)
print("Squared prices:", squared)
print("Total revenue sum:", np.sum(total))
```

We created two arrays containing product prices and quantities. We multiplied the arrays element by element to calculate revenue for each product. We then subtracted 10 from every price, squared the prices using the power operator, and used np.sum() to calculate the combined revenue of all products.

## Universal Functions (ufuncs)

Universal Functions (ufuncs) are NumPy functions designed to perform operations element by element on arrays efficiently. They include mathematical functions such as square root, exponential, sine, and logarithm, along with arithmetic and comparison operations.
### Formula
For the square root function:
\[
y_i=\sqrt{x_i}
\]

Each element in the input array is transformed independently.
### Numerical Example
Input:
[1, 4, 9, 16, 25]
Square roots:
[1, 2, 3, 4, 5]
### Where It Is Used
- Applying mathematical functions to numerical datasets.
- Transforming skewed features.
- Performing scientific calculations.
- Preparing numerical features for machine learning models.
### AI/ML Engineer Example
An AI/ML engineer applies a logarithmic transformation to highly skewed sales values to reduce their skewness and make the data easier to analyze. The transformation used should depend on the data and the model's requirements.

```python
import numpy as np

arr = np.array([1, 4, 9, 16, 25])

print("Square root:", np.sqrt(arr))
print("Exponential:", np.exp(np.array([0, 1, 2])))
print("Logarithm:", np.log(np.array([1, np.e, np.e**2])))
print("Absolute values:", np.abs(np.array([-10, -5, 5, 10])))
```

We created arrays and applied several NumPy universal functions. np.sqrt() calculated the square root of every element, np.exp() calculated the exponential of each value, and np.log() calculated the natural logarithm. We also used np.abs() to convert negative values into their absolute values. These functions perform element-wise transformations without requiring manual loops.

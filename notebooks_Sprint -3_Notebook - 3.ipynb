# NumPy Random Module

## Random Numbers — np.random.random()

Random numbers are values generated through a process that makes them unpredictable in advance. Computers commonly generate pseudorandom numbers, which are produced by mathematical algorithms but behave like random values for many practical applications.
NumPy provides functions to generate random floating-point numbers. The np.random.random() function generates values from a uniform distribution over the interval from zero, inclusive, to one, exclusive. This means every value in that interval has the same probability density.
### Formula
For a uniform random variable over the interval from zero to one:
\[
X \sim U(0,1)
\]

Its probability density is:
\[
f(x)=
\begin{cases}
1, & 0\leq x<1\\
0, & \text{otherwise}
\end{cases}
\]

### Numerical Example
Suppose we generate three random values. One possible result is:
[0.23, 0.71, 0.48]


These are example values; your program may produce different numbers.
### Where It Is Used
- Running simulations and experiments.
- Generating random test data.
- Initializing numerical calculations.
- Creating randomized samples for statistical analysis.
### AI/ML Engineer Example
An AI/ML engineer can use random numbers to initialize model parameters or generate synthetic numerical inputs for testing a data-processing pipeline.

```python
import numpy as np

random_values = np.random.random(5)

print("Random Numbers:", random_values)
```

We used np.random.random(5) to generate five random floating-point values between zero and one. The argument specifies how many values to generate. We stored the result in random_values and printed the array. Because the numbers are pseudorandom, running the code again will generally produce a different sequence unless the random generator's state is controlled.

## Random Seed — np.random.seed()

A random seed is an initial value used to initialize a pseudorandom number generator. The generator uses this starting state to produce a sequence of values. When the same seed is used with the same generator, the same sequence can be reproduced, provided the generation process and software environment are compatible.
Random seeds are especially important in machine learning because many algorithms involve random initialization, random data splitting, or random sampling. Fixing a seed makes experiments easier to reproduce and compare. A seed does not make numbers more random; it makes the sequence reproducible.
### Formula
There is no single universal seed formula. Conceptually, a pseudorandom generator updates its internal state using a deterministic rule:
\[
S_{t+1}=F(S_t)
\]

The generator then produces an output from its state. The function \(F\) depends on the algorithm being used.
### Numerical Example
Using the same seed twice produces the same sequence:
First run:  [0.37, 0.95, 0.73]
Second run: [0.37, 0.95, 0.73]


The values above are illustrative; the actual sequence depends on the generator and seed.
### Where It Is Used
- Reproducing machine learning experiments.
- Comparing models under consistent conditions.
- Creating repeatable test data.
- Debugging code that involves randomness.
### AI/ML Engineer Example
An AI/ML engineer sets a random seed before initializing a model so that repeated experiments can begin with the same random initialization. This makes comparisons easier, although it does not guarantee complete reproducibility across every framework, device, or hardware configuration.

```python
import numpy as np

np.random.seed(42)
first_run = np.random.random(3)

np.random.seed(42)
second_run = np.random.random(3)

print("First Run:", first_run)
print("Second Run:", second_run)
print("Same Sequence:", np.array_equal(first_run, second_run))
```

We set the seed to 42 using np.random.seed(42) and generated three random values. We then reset the seed to the same value and generated another three values. Because the generator started from the same state and performed the same operations, both arrays contain identical values. Finally, np.array_equal() checked whether the arrays matched and returned True. For newer NumPy code, the recommended approach is to create a generator using np.random.default_rng(seed).

## Random Integers — np.random.randint()

Random integers are whole-number values generated within a specified range. NumPy's np.random.randint() function selects integers from a range, including the lower limit but excluding the upper limit. You can generate a single integer or an array of integers by specifying the desired size.
When generating integers without additional weighting, the possible integer values are selected uniformly. This means each allowed integer has the same probability of being selected.
### Formula
For integers selected uniformly from the range \([a,b)\), the probability of each allowed integer is:
\[
P(X=k)=\frac{1}{b-a}
\]

where \(a\) is the inclusive lower limit and \(b\) is the exclusive upper limit.
### Numerical Example
Suppose we generate five random integers between 1 and 10, excluding 10.
One possible output is:
[3, 7, 1, 9, 4]


### Where It Is Used
- Generating random IDs for test data.
- Simulating dice rolls.
- Creating random integer datasets.
- Selecting random indices from arrays.
### AI/ML Engineer Example
An AI/ML engineer generates random integer indices to select samples for testing a data-processing pipeline or to simulate discrete input values for a classification experiment.

```python
import numpy as np

rng = np.random.default_rng(42)

random_integers = rng.integers(1, 10, size=5)

print("Random Integers:", random_integers)
```

We created a random number generator using np.random.default_rng(42). The rng.integers(1, 10, size=5) method generated five integers from 1 through 9, because the upper limit 10 is excluded. We stored these values in random_integers and printed them. Setting the seed makes the output repeatable when the same generator and sequence of operations are used.

## Random Choice — np.random.choice()
Definition
Random choice selects one or more elements from a given array or collection. It can select elements with replacement, meaning the same element may be selected multiple times, or without replacement, meaning an element cannot be selected more than once in the same selection.
NumPy also allows you to specify probabilities for individual elements. This makes it possible to model situations where some outcomes are more likely than others.
Formula
When all \(n\) possible elements are equally likely to be selected:
\[
P(X=x_i)=\frac{1}{n}
\]

When probabilities are specified, each element can have its own probability, provided the probabilities are valid and sum to one.

### Where It Is Used
- Randomly selecting records from a dataset.
- Sampling products or customers.
- Creating randomized training examples.
- Simulating events with different possible outcomes.
### AI/ML Engineer Example
An AI/ML engineer randomly selects a subset of records from a large dataset to test a data transformation or inspect sample predictions. If the dataset is imbalanced, probabilities can also be used to change the sampling proportions.

```python
import numpy as np

rng = np.random.default_rng(42)

colors = np.array(["Red", "Green", "Blue"])

selected_colors = rng.choice(colors, size=2, replace=False)

print("Selected Colors:", selected_colors)
```

We created an array containing three colors and used rng.choice() to select two of them randomly. The size=2 argument specifies how many elements to select, while replace=False prevents the same color from being selected twice. We stored the selected values in selected_colors and printed them. This method is useful for random sampling when duplicate selections are not allowed.

### Shuffle — np.random.shuffle()

Shuffling rearranges the elements of an array into a random order. NumPy's shuffle() function modifies the original array in place, meaning that the array itself changes rather than returning a separately shuffled array.
For a one-dimensional array, the elements are rearranged randomly. For a multidimensional array, shuffle() rearranges the elements along the first axis, so entire rows are shuffled rather than individual values inside every row.
### Formula
There is no arithmetic formula required for shuffling. Conceptually, it generates a random permutation of the elements.
For an array of distinct elements, there are:
\[
n!
\]

possible orderings, where \(n\) is the number of elements. An unbiased shuffle can select each ordering with equal probability.

### Where It Is Used
- Randomizing the order of training data.
- Preventing models from learning unwanted ordering patterns.
- Creating randomized test sequences.
- Mixing records before sampling or splitting data.
### AI/ML Engineer Example
An AI/ML engineer shuffles training examples before each training epoch so that the model does not repeatedly process examples in the same order. This is particularly useful when the original dataset is sorted by class, date, or source.

```python
import numpy as np

rng = np.random.default_rng(42)

data = np.array([10, 20, 30, 40, 50])

print("Original Array:", data)

rng.shuffle(data)

print("Shuffled Array:", data)
```

We created a NumPy array named data and printed its original order. Then, rng.shuffle(data) rearranged its elements in place using the random generator. Because the operation modifies the original array, printing data afterward displays the shuffled order. The exact order is determined by the generator's seed and the operations performed before the shuffle.

## Permutation — np.random.permutation()

A permutation is a rearrangement of elements into a different order. NumPy's permutation() function returns a randomly permuted sequence or array without modifying the original input array.
It can also generate a permutation of integer indices. This is useful when you want to randomize the order of records while preserving the original dataset. Unlike shuffling, which modifies an existing array, permutation produces a separate result.
### Formula
The number of possible permutations of \(n\) distinct elements is:
\[
P(n)=n!
\]

A permutation changes the order of elements but does not add or remove any elements.

### Where It Is Used
- Generating random index orders.
- Creating randomized datasets without changing the source.
- Preparing data for training and testing.
- Testing algorithms with different input orderings.
### AI/ML Engineer Example
An AI/ML engineer creates a random permutation of dataset indices and uses those indices to reorder features and labels consistently before training a model.

```python
import numpy as np

rng = np.random.default_rng(42)

data = np.array([10, 20, 30, 40])

permuted_data = rng.permutation(data)

print("Original Array:", data)
print("Permuted Array:", permuted_data)
```

We created an original array and passed it to rng.permutation(). NumPy returned a new array with the same elements arranged in a randomized order. We stored the result in permuted_data, allowing us to print both arrays and confirm that the original array remained unchanged. This is helpful when we need a randomized copy while keeping the source data intact.

## Random Distributions

A probability distribution describes how likely different values or outcomes are when a random process occurs. Random distributions allow us to generate values that follow a particular probability pattern instead of selecting values uniformly from a range.
NumPy supports many distributions, including uniform, normal, binomial, and Poisson distributions. Each distribution models a different type of real-world situation. Choosing the correct distribution helps engineers simulate realistic data, test statistical assumptions, and understand how random variation affects a system.
### Formula
A probability distribution describes the probabilities associated with possible outcomes. The specific formula depends on the distribution being used.
For example, a normal distribution is described by its mean and standard deviation:
\[
X\sim\mathcal{N}(\mu,\sigma^2)
\]

Here, \(\mu\) represents the mean and \(\sigma\) represents the standard deviation.
### Numerical Example
Suppose we generate random values from a normal distribution representing measurement errors. One possible set of values is:
[-0.42, 0.18, 1.05, -0.73, 0.31]


These values are illustrative; actual values depend on the distribution parameters and random generator.
### Where It Is Used
- Simulating real-world measurements and random events.
- Generating synthetic datasets.
- Testing statistical models.
- Modeling uncertainty and variability.
- Creating random samples for machine learning experiments.
### AI/ML Engineer Example
An AI/ML engineer generates synthetic data from a normal distribution to test a preprocessing pipeline or simulate measurement noise before applying the pipeline to real data.

```python
import numpy as np

rng = np.random.default_rng(42)

normal_data = rng.normal(loc=0, scale=1, size=5)

print("Normal Distribution:", normal_data)
```

We created a random generator and used rng.normal() to generate five values from a normal distribution. The loc=0 argument sets the mean, scale=1 sets the standard deviation, and size=5 specifies the number of values to generate. The values tend to cluster around the mean, although some values may be farther away. We stored the results in normal_data and printed them. The seed ensures that this sequence can be reproduced under the same conditions.

# Combining Datasets 
Combining datasets in Python means bringing data from multiple DataFrames together into a single DataFrame. Pandas provides methods such as concat(), merge(), join(), and combine_first() to combine data vertically, horizontally, or using common keys.

```python
%mamba install pandas
```

```python
import pandas as pd

employees1 = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Arun"],
    "Department": ["IT", "HR", "Finance"]
})

employees2 = pd.DataFrame({
    "Employee_ID": [104, 105, 106],
    "Name": ["Sneha", "Kiran", "Manoj"],
    "Department": ["IT", "HR", "Finance"]
})

salaries = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Salary": [30000, 45000, 40000, 35000]
})

print(employees1)
print(salaries)
```

## concat()

The concat() function combines two or more pandas objects by stacking them vertically or horizontally. It aligns data using row or column labels, depending on the specified axis.
### Formula 
pd.concat([df1, df2], axis=0)

- axis=0: Combines rows vertically.
- axis=1: Combines columns horizontally.
No mathematical formula is required.
### When it is used
- To combine data collected from multiple files.
- To append records from different months or departments.
- To combine DataFrames with the same or compatible columns.
### AI/ML example
You have separate customer datasets for January and February. You can use concat() to combine the records into one dataset before analyzing customer behavior.

```python
# Combine employee datasets vertically
result = pd.concat(
    [employees1, employees2],
    axis=0,
    ignore_index=True
)

print(result)
```

We used pd.concat() to combine employees1 and employees2 vertically because both DataFrames contain the same columns. The axis=0 parameter places the rows from the second DataFrame below the rows from the first DataFrame. We also used ignore_index=True to create a fresh sequential index from 0 to 5. This is useful when combining records from different files that share the same structure.

## merge()

The merge() method combines two DataFrames by matching values in one or more key columns. It supports SQL-style joins such as inner, left, right, and outer joins.
### Formula 
pd.merge(df1, df2, on="common_column", how="inner")
### Common join types:
- inner: Keeps matching keys from both DataFrames.
- left: Keeps all keys from the left DataFrame.
- right: Keeps all keys from the right DataFrame.
- outer: Keeps all keys from both DataFrames.
### When it is used
- To combine related datasets using a common identifier.
- To connect customer records with transaction details.
- To combine employee information with salary or attendance data.
### AI/ML example
In a customer churn prediction project, you may merge customer profile data with monthly billing data using Customer_ID to build a more complete dataset.

```python
# Merge employee details with salary information
result = pd.merge(
    employees1,
    salaries,
    on="Employee_ID",
    how="inner"
)

print(result)
```

We used pd.merge() to combine employee details with salary information using Employee_ID as the common key. The how="inner" parameter keeps only employee IDs present in both DataFrames. Since employees 101, 102, and 103 appear in both datasets, their names, departments, and salaries are combined into one table. Employee 104 is excluded because it is not present in employees1.

## join()

The join() method combines DataFrames using their index labels by default. It can also join using a column from the calling DataFrame and the index of the other DataFrame.
### Formula 
df1.join(df2, how="left")
By default, join() uses a left join based on the index.
### When it is used
- To combine DataFrames that already share meaningful index labels.
- To attach additional features to an existing dataset.
- To join indexed time-series or panel data.
### AI/ML example
In a time-series forecasting project, you may join weather information with sales data when both datasets use the same date as their index.

```python
employee_info = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"]
}, index=[101, 102, 103])

salary_info = pd.DataFrame({
    "Salary": [30000, 45000, 40000]
}, index=[101, 102, 103])

result = employee_info.join(salary_info)

print(result)
```

We created two DataFrames that use employee IDs as their index labels. The first contains employee names, and the second contains salaries. We used join() to combine the two DataFrames by matching their index labels. Since both DataFrames contain the same indexes, each employee receives the correct salary. This is particularly convenient when datasets are already indexed by a shared identifier.

## append() — Deprecation and alternatives

Historically, the DataFrame.append() method added rows from one DataFrame to another and returned a new DataFrame. It was deprecated in pandas 1.4 and removed in pandas 2.0, so modern pandas code should use pd.concat() instead.
### Formula 
Old syntax — do not use in modern pandas:
Deprecated and removed in pandas 2.0
result = df1.append(df2, ignore_index=True)
Modern alternative:
result = pd.concat([df1, df2], ignore_index=True)
No mathematical formula is required.
### When it is used
- When updating older pandas code that uses append().
- When combining rows from multiple DataFrames.
- When maintaining compatibility with pandas 2.0 and newer versions.
### AI/ML example
A data pipeline may collect training records from several sources. Replacing old append() calls with pd.concat() keeps the pipeline compatible with modern pandas.

```python
df1 = pd.DataFrame({
    "Name": ["Ravi", "Priya"],
    "Salary": [30000, 45000]
})

df2 = pd.DataFrame({
    "Name": ["Arun", "Sneha"],
    "Salary": [40000, 28000]
})

# Modern replacement for append()
result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)
```

We created two DataFrames containing employee names and salaries, then combined them using pd.concat(). The ignore_index=True parameter generates a new sequential index for the combined DataFrame. This produces the same kind of row-stacking result that developers commonly used append() for in older pandas code. Because append() was removed in pandas 2.0, concat() is the recommended alternative.

## combine_first()

The combine_first() method fills missing values in one DataFrame or Series using corresponding non-missing values from another. It aligns data by index and, for DataFrames, column labels, while preserving existing non-missing values in the calling object.
### Formula 
df1.combine_first(df2)

The basic rule is:
- If df1 has a value, keep it.
- If df1 has a missing value, use the corresponding value from df2 if available.
### When it is used
- To fill missing values using another dataset.
- To combine overlapping datasets with partially complete information.
- To reconcile records from different sources while preserving existing values.
### AI/ML example
In a customer dataset, one source may have missing income values while another source has those incomes recorded. combine_first() can fill the missing values before further preprocessing, provided that the records and indexes are correctly aligned.

```python
df1 = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Salary": [30000, None, 40000]
}, index=[101, 102, 103])

df2 = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Salary": [32000, 45000, 42000]
}, index=[101, 102, 103])

result = df1.combine_first(df2)

print(result)
```

We created two DataFrames containing employee information, where Priya's salary is missing in the first DataFrame. The second DataFrame contains a salary for every employee. We used combine_first() to preserve Ravi's existing salary of ₹30,000 and Arun's existing salary of ₹40,000, while filling Priya's missing salary with ₹45,000 from the second DataFrame. This method is useful when combining partially complete records, but it depends on correct index and column alignment.

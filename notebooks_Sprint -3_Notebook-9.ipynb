# Grouping & Aggregation 
Grouping and aggregation in Python means organizing data into categories and calculating summary information for each category. In pandas, these operations help us analyze large datasets by department, region, product, customer type, or other features, and they are commonly used in Data Analysis and Machine Learning.

```python
%mamba install pandas
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Manoj"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Male"],
    "Age": [25, 30, 28, 24, 35, 27],
    "Salary": [30000, 45000, 40000, 28000, 60000, 35000]
}

df = pd.DataFrame(data)

print(df)
```

## groupby()

The groupby() method divides a DataFrame into groups based on one or more columns. It allows us to perform calculations such as sum, mean, count, minimum, and maximum separately for each group.
### Formula
For calculating the average of each group:
\[
\text{Group Mean}=\frac{\text{Sum of values in group}}{\text{Number of values in group}}
\]

### Numerical example
Consider the salaries of employees in the IT department:
30000, 40000, 35000
\[
\text{Average Salary}=\frac{30000+40000+35000}{3}=35000
\]

Therefore, the average salary in the IT department is ₹35,000.
### When it is used
- To divide records into categories for analysis.
- To calculate statistics for different groups.
- To compare performance across departments, regions, or categories.
### AI/ML example
In a customer churn prediction project, groupby() can calculate the average monthly spending of customers who churned versus those who stayed. This helps identify potential relationships between customer spending and churn.

```python
# Group employees by department and calculate average salary
result = df.groupby("Department")["Salary"].mean()

print(result)
```

We used the groupby() method to divide employees into groups based on their departments. After grouping, we selected the Salary column and applied the mean() method to calculate the average salary for each department. The output shows that Finance has an average salary of ₹28,000, HR has ₹52,500, and IT has ₹35,000. This helps us compare salary distributions across different departments without calculating each group's average manually.

## agg()

The agg() method performs one or more aggregation operations on a DataFrame or grouped data. It allows us to calculate multiple statistics, such as mean, sum, minimum, maximum, and count, in a single operation.
### Formula
For calculating the total value of a group:
\[
\text{Group Total}=\sum_{i=1}^{n}x_i
\]

### Numerical example
Consider the salaries in the HR department:
45000, 60000
\[
\text{Total Salary}=45000+60000=105000
\]

Therefore, the total salary of HR employees is ₹1,05,000.
### When it is used
- To calculate multiple summary statistics at once.
- To create department-wise or category-wise reports.
- To summarize large datasets efficiently.
### AI/ML example
In a sales forecasting project, agg() can calculate the total sales, average transaction value, and number of transactions for each store. These statistics can help identify useful patterns and create aggregated features.

```python
# Calculate multiple statistics for each department
result = df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max", "count"]
)

print(result)
```

We first grouped employees according to their departments and then used agg() to calculate five different statistics for their salaries. The sum operation calculates the total salary, mean calculates the average, min and max identify the lowest and highest salaries, and count counts the non-missing salary values. Instead of writing separate statements for each calculation, agg() produces all the results in a single table, making grouped analysis more efficient.

## transform()

The transform() method performs calculations on groups and returns results aligned with the original rows. Unlike aggregation methods that reduce each group to a summary, transform() preserves the original number of rows.
### Formula
For calculating the difference between an individual salary and the department average:
\[
\text{Salary Difference}=\text{Employee Salary}-\text{Department Average Salary}
\]

### Numerical example
Suppose an employee earns ₹40,000 and the average salary of their department is ₹35,000.
\[
\text{Salary Difference}=40000-35000=5000
\]

The employee earns ₹5,000 more than the department average.
### When it is used
- To create group-based features.
- To compare individual values with group averages.
- To normalize or standardize values within groups.
### AI/ML example
In an employee attrition prediction model, transform() can create a feature showing how much an employee's salary differs from their department's average salary. Such a feature may help a model identify salary-related patterns, provided group statistics are calculated without data leakage.

```python
# Calculate department average salary for each employee
df_transformed = df.copy()

df_transformed["Department_Avg_Salary"] = (
    df_transformed.groupby("Department")["Salary"]
    .transform("mean")
)

print(df_transformed[
    ["Name", "Department", "Salary", "Department_Avg_Salary"]
])
```

We created a copy of the employee DataFrame and grouped the records by department. We then used transform("mean") to calculate the average salary for each department and assign that average to every employee belonging to the corresponding department. For example, Ravi, Arun, and Manoj all receive ₹35,000 as their department average because they belong to IT. Unlike groupby().mean(), which produces one result per department, transform() maintains the original number of rows, making it useful for creating new features.

## filter()

The filter() method removes entire groups that do not satisfy a specified condition. When used with groupby(), it evaluates a condition for each group and retains all rows belonging to groups that satisfy that condition.
### Formula
For filtering groups based on their size:
\[
\text{Keep Group if Group Count}\geq k
\]

Here, \(k\) represents the minimum number of records required in a group.
### Numerical example
Suppose three departments contain:
- IT: 3 employees
- HR: 2 employees
- Finance: 1 employee
If the minimum group size is 2, IT and HR are retained, while Finance is removed.
### When it is used
- To remove groups with insufficient records.
- To retain groups that meet specific business conditions.
- To focus analysis on categories with enough observations.
### AI/ML example
In a product recommendation dataset, filter() can retain products that have received at least five ratings. This can help focus an analysis on products with sufficient feedback, although it may also introduce selection bias.

```python
# Keep departments having at least 2 employees
result = df.groupby("Department").filter(
    lambda group: len(group) >= 2
)

print(result)
```

We grouped the employee DataFrame by department and used filter() with a lambda function to check the number of employees in each group. The expression len(group) >= 2 keeps departments containing at least two employees. Since IT contains three employees and HR contains two, both departments are retained. Finance is removed because it contains only one employee. The method returns all original rows belonging to the retained groups rather than returning just a summary for each group.

## Pivot Tables

A Pivot Table summarizes and reorganizes data by grouping values across rows and columns. In pandas, pivot_table() can calculate statistics such as sum, mean, and count for combinations of categories.
### Formula
For calculating the average salary in each category:
\[
\text{Average Salary}=\frac{\text{Total Salary}}{\text{Number of Employees}}
\]

### Numerical example
Consider two employees in the HR department with salaries of ₹45,000 and ₹60,000.
\[
\text{Average Salary}=\frac{45000+60000}{2}=52500
\]

The average HR salary is ₹52,500.
### When it is used
- To summarize data across multiple categories.
- To compare values using rows and columns.
- To create business reports and analytical summaries.
### AI/ML example
In a retail sales prediction project, a Pivot Table can calculate average monthly sales for each store and product category. This helps identify seasonal or category-level patterns that may be useful for forecasting.

```python
# Create a pivot table of average salary
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean",
    fill_value=0
)

print(pivot)
```

We created a Pivot Table using pd.pivot_table() to calculate average salaries based on department and gender. The values parameter specifies that salaries should be analyzed, index places departments in rows, and columns places genders in separate columns. The aggfunc="mean" parameter calculates the average salary for each combination, while fill_value=0 replaces missing combinations with zero. These zeros indicate combinations with no observations, not actual zero salaries. This structure makes it easier to compare salaries across multiple categories.

## Crosstab

A Crosstab displays the frequency distribution of observations across two or more categorical variables. In pandas, pd.crosstab() creates a table showing how often combinations of categories occur.
### Formula
For calculating the percentage of observations in a category:
\[
\text{Percentage}=\frac{\text{Category Count}}{\text{Total Count}}\times100
\]

### Numerical example
Suppose a company has 10 employees, and 4 belong to the IT department.
\[
\text{IT Percentage}=\frac{4}{10}\times100=40\%
\]

Therefore, 40% of the employees belong to IT.
### When it is used
- To count combinations of categorical values.
- To understand relationships between categorical variables.
- To analyze class distributions and category frequencies.
### AI/ML example
In a customer churn prediction project, a Crosstab can compare customer churn status across different contract types. This helps identify whether churn rates differ between monthly and annual contracts.

```python
# Count employees by department and gender
result = pd.crosstab(
    df["Department"],
    df["Gender"]
)

print(result)
```

We used pd.crosstab() to count employees based on department and gender. The departments appear as rows, while Male and Female appear as columns. Each cell represents the number of employees belonging to that combination. For example, the IT department contains three male employees and no female employees, while HR contains one male and one female employee. Crosstabs are useful for understanding categorical distributions and identifying patterns before applying statistical tests or building classification models.

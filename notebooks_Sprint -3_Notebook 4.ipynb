# Introduction to Pandas

## What Is Pandas?

Pandas is an open-source Python library used to work with structured and tabular data. It provides powerful data structures, such as Series and DataFrame, for organizing, manipulating, and analyzing data.
### Formula
Not applicable. Pandas is a software library, not a mathematical operation.
### Numerical Example
Suppose we have student data containing marks:
Student	Marks
Ravi	80
Priya	90
Kiran	75
Pandas allows us to store this data in a structured format and calculate statistics such as average marks.
### Where It Is Used
- Data analysis and reporting.
- Cleaning missing or incorrect data.
- Reading CSV and Excel files.
- Preparing datasets for machine learning.
### AI/ML Engineer Example
An AI/ML engineer uses Pandas to load customer data, inspect its columns, identify missing values, and prepare the dataset before training a machine learning model.

```python
import pandas as pd

data = {
    "Student": ["Ravi", "Priya", "Kiran"],
    "Marks": [80, 90, 75]
}

df = pd.DataFrame(data)

print(df)
```

## Why Pandas?

Pandas makes working with structured data easier by providing convenient functions for filtering, sorting, grouping, and cleaning datasets. It also integrates with libraries such as NumPy, Matplotlib, and Scikit-learn.
### Formula
Not applicable. Pandas provides data manipulation tools rather than a single mathematical formula.
### Numerical Example
Suppose a shop records the following sales:
Product	Sales
Laptop	50000
Phone	20000
Tablet	15000
Pandas can quickly identify products with sales above 20000 and calculate the total sales.
### Where It Is Used
- Filtering rows based on conditions.
- Calculating summary statistics.
- Sorting and grouping data.
- Preparing data for visualizations and machine learning.
### AI/ML Engineer Example
An AI/ML engineer uses Pandas to filter customer records, remove duplicates, and select relevant features before passing the data to a model.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50000, 20000, 15000]
})

high_sales = df[df["Sales"] > 20000]

print(high_sales)
print("Total Sales:", df["Sales"].sum())
```

We created a DataFrame containing product names and sales amounts. The expression df["Sales"] > 20000 checks each row and returns a Boolean condition, which we used to select only products meeting that threshold. We then used df["Sales"].sum() to calculate total sales across all rows. This shows how Pandas combines filtering and calculations in a small amount of code.

## Series

A Series is a one-dimensional Pandas data structure that stores values along with an index. It is similar to a single column in a table and can contain numbers, strings, or other data types.
### Formula
Not applicable. A Series is a data structure rather than a mathematical formula.
### Numerical Example
Consider a Series of marks:
0    80
1    90
2    75


Each value has an index that identifies its position or label.
### Where It Is Used
- Storing one column of data.
- Performing calculations on individual columns.
- Filtering and selecting values.
- Analyzing a single variable.
### AI/ML Engineer Example
An AI/ML engineer uses a Series to store a target variable, such as house prices, or to examine a single feature, such as customer age.

```python
import pandas as pd

marks = pd.Series([80, 90, 75])

print(marks)
print("Average Marks:", marks.mean())
```

We created a Series using pd.Series() and supplied a list of marks. Pandas automatically assigned an integer index starting at zero because we did not specify custom labels. The mean() method calculated the average of the values, and printing the Series displayed both the indexes and their corresponding marks. This is useful when we want to analyze a single column of data independently.

## DataFrame

A DataFrame is a two-dimensional Pandas data structure arranged into rows and columns, similar to a spreadsheet or database table. Each column can have its own data type, and rows can be identified using an index.
### Formula
Not applicable. A DataFrame is a tabular data structure.
### Numerical Example
Name	Age	Salary
Ravi	25	30000
Priya	28	45000
Kiran	24	28000
This table contains multiple related attributes for each person.
### Where It Is Used
- Storing structured datasets.
- Combining related columns.
- Filtering, sorting, and grouping records.
- Preparing datasets for reporting and machine learning.
### AI/ML Engineer Example
An AI/ML engineer stores customer features such as age, income, and purchase history in a DataFrame before selecting the features needed for model training.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Kiran"],
    "Age": [25, 28, 24],
    "Salary": [30000, 45000, 28000]
})

print(df)
```

We created a dictionary where each key represents a column name and each list contains that column's values. Passing this dictionary to pd.DataFrame() created a table with three columns and three rows. Pandas assigned default integer indexes to the rows. We then printed the DataFrame to view all the information together, which is the typical starting point for data analysis.

## Creating DataFrames

Creating a DataFrame means converting data from Python structures or external sources into a table with rows and columns. Pandas supports creating DataFrames from dictionaries, lists, NumPy arrays, and other tabular data sources.
### Formula
Not applicable. DataFrame creation is a data-structuring operation.
### Numerical Example
Suppose we have three employees with their IDs and salaries:
Employee ID	Salary
101	30000
102	40000
103	35000
We can create a DataFrame from these values.
### Where It Is Used
- Creating datasets from manually collected information.
- Converting API responses into tables.
- Organizing numerical arrays into named columns.
- Building sample datasets for testing code.
### AI/ML Engineer Example
An AI/ML engineer creates a DataFrame from model predictions and actual outcomes to compare results and evaluate model performance.

```python
import pandas as pd

data = [
    {"Employee_ID": 101, "Salary": 30000},
    {"Employee_ID": 102, "Salary": 40000},
    {"Employee_ID": 103, "Salary": 35000}
]

df = pd.DataFrame(data)

print(df)
```

We created a list of dictionaries, with each dictionary representing one employee's record. We passed this list to pd.DataFrame() to create a table in which the dictionary keys became column names and the dictionary values became row data. Pandas automatically generated the row indexes. This method is useful when data is received as individual records from an API or another Python application.

## Data Types

Data types describe the kind of values stored in a Pandas Series or DataFrame column. Pandas supports types such as integers, floating-point numbers, strings, Boolean values, dates, and specialized nullable types.
### Formula
Not applicable. Data types define how values are represented and processed.
### Numerical Example
Consider this dataset:
Name	Age	Salary	Active
Ravi	25	30000.5	True
Priya	28	45000.0	False
Here, Name contains text, Age contains integers, Salary contains decimal values, and Active contains Boolean values.
### Where It Is Used
- Checking the structure of a dataset.
- Identifying incorrect data types.
- Converting text-based numbers into numerical values.
- Preparing clean data for mathematical calculations and machine learning.
### AI/ML Engineer Example
An AI/ML engineer checks column data types before training a model to ensure that numerical features are stored as numbers instead of text and that date columns are parsed correctly.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya"],
    "Age": [25, 28],
    "Salary": [30000.5, 45000.0],
    "Active": [True, False]
})

print(df.dtypes)
```

We created a DataFrame with text, integer, floating-point, and Boolean columns. The dtypes attribute displayed the data type Pandas inferred for each column. The text column appears as object in this example, although modern Pandas also supports dedicated string types. Checking data types helps us catch problems before performing calculations or training machine learning models.


## Index

An index identifies the rows of a Pandas Series or DataFrame and allows values to be selected by labels. Pandas creates a default integer index when one is not specified, but an index can also contain custom labels such as employee IDs or dates.
### Formula
Not applicable. An index is a labeling and selection mechanism, not a mathematical operation.
### Numerical Example
Suppose employee names are indexed by employee IDs:
Index (Employee ID)	Name
101	Ravi
102	Priya
103	Kiran
The index identifies each record using its employee ID.
### Where It Is Used
- Selecting specific rows by label.
- Identifying records using unique keys.
- Aligning data from multiple Series or DataFrames.
- Working with time-series data indexed by dates.
### AI/ML Engineer Example
An AI/ML engineer uses an index to align actual labels with model predictions so that each prediction is compared with the correct record.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Kiran"],
    "Salary": [30000, 45000, 35000]
}, index=[101, 102, 103])

print(df)
print("\nEmployee 102:")
print(df.loc[102])
```

We created a DataFrame and specified employee IDs as its index. This replaced the default integer index with meaningful labels. We then used df.loc[102] to retrieve the row whose index label is 102, returning Priya's details. The index helps us identify and select records without relying solely on their physical row positions.

## Columns

Columns represent individual variables or attributes in a DataFrame, with each column containing values for that variable across the rows. Pandas allows us to inspect, select, rename, add, and remove columns.
### Formula
Not applicable. Columns organize data into named variables.
### Numerical Example
Consider a DataFrame with these columns:
Name	Age	Salary
Ravi	25	30000
Priya	28	45000
Here, Name, Age, and Salary are column names.
### Where It Is Used
- Selecting features for analysis.
- Renaming unclear column names.
- Creating new calculated columns.
- Removing irrelevant or duplicate features.
### AI/ML Engineer Example
An AI/ML engineer selects columns such as age, salary, and purchase frequency as model features while keeping a separate column for the target variable.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya"],
    "Age": [25, 28],
    "Salary": [30000, 45000]
})

print("Column Names:", df.columns.tolist())

print("\nAge Column:")
print(df["Age"])

df["Bonus"] = df["Salary"] * 0.10

print("\nUpdated DataFrame:")
print(df)
```

We created a DataFrame containing employee names, ages, and salaries. The df.columns.tolist() expression retrieved the column names as a Python list, while df["Age"] selected the Age column as a Series. We then created a new column named Bonus by calculating 10% of each employee's salary. Printing the updated DataFrame showed how Pandas allows us to inspect existing columns and add new calculated features.

## Writing Data

Writing data means exporting a Pandas Series or DataFrame to an external file so that the results can be stored, shared, or used by another application. Pandas supports writing data to formats such as CSV, Excel, and JSON.
### Formula
Not applicable. Writing data is an output operation.
### Numerical Example
Suppose we have this DataFrame:
Name	Salary
Ravi	30000
Priya	45000
We can export it to a CSV file named employee_salaries.csv.
### Where It Is Used
- Saving cleaned datasets.
- Exporting analysis results.
- Sharing processed data with other teams.
- Saving outputs from data-processing pipelines.
### AI/ML Engineer Example
An AI/ML engineer exports a cleaned dataset to a CSV file so that another script or team member can reuse the prepared data without repeating the cleaning steps.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Priya"],
    "Salary": [30000, 45000]
})

df.to_csv("employee_salaries.csv", index=False)

print("Data saved successfully!")
```

We created a DataFrame containing employee names and salaries, then used df.to_csv() to export it to a CSV file. The argument index=False prevents Pandas from adding the DataFrame's row index as an extra column in the output file. Finally, we printed a confirmation message. In a real workflow, this lets us save cleaned or transformed data for later use by other programs or machine learning pipelines.

Other useful writing functions:
Function	Purpose
df.to_csv()	Write a CSV file
df.to_excel()	Write an Excel file
df.to_json()	Write a JSON file
df.to_sql()	Write data to a SQL table
Note that to_excel() may require an Excel-writing library such as openpyxl, depending on the file format and environment.
